### Problem 1: Quadratic Equation (Gezerlis Problem 2.6)

Take the standard quadratic equation: $ax^2 + bx + c = 0$, whose solutions are 

$$x_{\pm}=\frac{-b \pm \sqrt{b^2-4ac}}{2a}$$

Take $b>0$ for concreteness. It is easy to see that when $b^2 >> ac$, we don't get a catastrophic cancellation when evaluating $b^2 - 4ac$ (we may still get a "benign" cancellation). Furthermore, $\sqrt{b^2-4ac} \approx b$. 

However, this means that $x_+$ will involve catastrophic cancellation in the numerator. We will employ an analytical trick in order to help us preserve significant figures. Observe that the product of the two roots must obey 

$$x_+ x_- = \frac{c}{a}$$

The answer now presents itself: use the standard quadratic formula to calculate $x_-$, for which no catastrophic cancellation takes place. Then, use $x_+ x_- = \frac{c}{a}$ to calculate $x_+$. Notice that you end up calculating $x_+$, via division only (i.e. without a catastrophic cancellation). 

Write a Python code (use `np.sqrt()`) that evaluates and prints out **(a)** $x_-$, **(b)** $x_+$ using the "bad formula", and **(c)** $x_+$ using the "good formula". 

Take $a=1$, $c=1$, $b=10^8$. Discuss the answers.

Lastly, **(d)** [FIGURE] Visualize the loss of precision by plotting the relative error of the "bad" $x_+$ formula. Vary $b$ using a log scaling (e.g., from $10^4$ to $10^9$). To highlight the effect of hardware limits, evaluate the bad formula using both 32-bit (np.float32) and 64-bit (np.float64) precision, comparing them against the "good" formula as the baseline.

```python
from math import log10
import matplotlib.pyplot as plt
import numpy as np

#constants
a=1
b=1e8
c=1

#looking if discriminant is imaginary
def calculate_discriminant(a,b,c):
    disc = b**2-4*a*c
    if disc<0:
        return False
    return disc

def minus_root(a,b,c):
    sqrt_discriminant = np.sqrt(calculate_discriminant(a,b,c))
    x_minus = -(b+sqrt_discriminant)/(2*a)
    return x_minus

def bad_plus_root(a,b,c):
    sqrt_discriminant = np.sqrt(calculate_discriminant(a,b,c))
    bad_x_plus = -(b-sqrt_discriminant)/(2*a)
    return bad_x_plus

def good_plus_root(a,b,c):
    good_x_plus = c / (a*minus_root(a,b,c))
    return good_x_plus

if calculate_discriminant(a,b,c) != False:
    item_1 = minus_root(a,b,c)
    item_2 = bad_plus_root(a,b,c)
    item_3 = good_plus_root(a,b,c)
    difference = abs(item_2 - item_3)
    err = abs(difference/item_3)
    print(f"minus root: {item_1}\nbad plus root: {item_2}\n good_plus_root: {item_3}\n\nDifference between plus roots{difference}\nError relative to good root{err}")
else:
    print("No real solutions")

b_values = np.logspace(1,9,500)
error_float32 = []
error_float64 = []

for b in b_values:
    a32 = np.float32(1.0)
    b32 = np.float32(b)
    c32 = np.float32(1.0)

    discriminant32 = b32**2 -np.float32(4.0)*a32*c32
    sqrt_discriminant32 = np.sqrt(discriminant32)

    x_bad32 = (-b+sqrt_discriminant32)/(np.float32(2.0)*a32)
    x_minus32 = (-b-sqrt_discriminant32)/(np.float32(2.0)*a32)
    x_good32 = c32/(a32*x_minus32)

    rel_error32 = abs((x_bad32-x_good32)/x_good32)
    error_float32.append(rel_error32)

    a64 = np.float64(1.0)
    b64 = np.float64(b)
    c64 = np.float64(1.0)

    discriminant64 = b64**2 - np.float64(4.0)*a64*c64
    sqrt_discriminant64 = np.sqrt(discriminant64)

    x_bad64 = (-b+sqrt_discriminant64)/(np.float64(2.0)*a64)
    x_minus64 = (-b-sqrt_discriminant64)/(np.float64(2.0)*a64)
    x_good64 = c64/(a64*x_minus64)

    rel_error64 = abs((x_bad64 - x_good64)/x_good64)
    error_float64.append(rel_error64)

plt.figure()
plt.loglog(
    b_values,
    error_float32,
    label = "32-bit (float32)",
    linewidth = 2
)

plt.loglog(
    b_values,
    error_float64,
    label = "64-bit (float64)",
    linewidth = 2
)

plt.xlabel("b")
plt.ylabel("Relative Error")
plt.title("Loss of precision in the 'bad' x formula")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ur code
import numpy as np
import matplotlib.pyplot as plt

def f_manual(q):
    return np.exp(q) - 1/q

def f_numpy(n):
    return np.expm1(n)/n

x = np.linspace(-16, 30, 50)

test2 = f_numpy(x)
test = f_manual(x)

plt.plot(test, label="Manual: (exp(x)-1)/x")
plt.plot(test2, label="Numpy: expm1(x)/x)")
plt.legend()
plt.axhline(1, linestyle="--", label="Analytical Limit = 1")
plt.xlabel('x')
plt.ylabel('y')
plt.title(r'Plot of $f(x) = \frac{e^x-1}{x}$')
plt.legend()
plt.grid()
plt.show()
```
