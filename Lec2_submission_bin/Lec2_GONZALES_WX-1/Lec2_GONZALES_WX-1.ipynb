### Problem 1: Quadratic Equation (Gezerlis Problem 2.6)

Take the standard quadratic equation: $ax^2 + bx + c = 0$, whose solutions are 

$$x_{\pm}=\frac{-b \pm \sqrt{b^2-4ac}}{2a}$$

Take $b>0$ for concreteness. It is easy to see that when $b^2 >> ac$, we don't get a catastrophic cancellation when evaluating $b^2 - 4ac$ (we may still get a "benign" cancellation). Furthermore, $\sqrt{b^2-4ac} \approx b$. 

However, this means that $x_+$ will involve catastrophic cancellation in the numerator. We will employ an analytical trick in order to help us preserve significant figures. Observe that the product of the two roots must obey 

$$x_+ x_- = \frac{c}{a}$$

The answer now presents itself: use the standard quadratic formula to calculate $x_-$, for which no catastrophic cancellation takes place. Then, use $x_+ x_- = \frac{c}{a}$ to calculate $x_+$. Notice that you end up calculating $x_+$, via division only (i.e. without a catastrophic cancellation). 

Write a Python code (use `np.sqrt()`) that evaluates and prints out **(a)** $x_-$, **(b)** $x_+$ using the "bad formula", and **(c)** $x_+$ using the "good formula". 

Take $a=1$, $c=1$, $b=10^8$. Discuss the answers.

Lastly, **(d)** [FIGURE] Visualize the loss of precision by plotting the relative error of the "bad" $x_+$ formula. Vary $b$ using a log scaling (e.g., from $10^4$ to $10^9$). To highlight the effect of hardware limits, evaluate the bad formula using both 32-bit (np.float32) and 64-bit (np.float64) precision, comparing them against the "good" formula as the baseline.

```python
import numpy as np
import matplotlib.pyplot as plt

def quad_roots(a, b, c):
    disc = b**2 - 4*a*c
    #identifies if the quadratic equation has real solutions

    if disc < 0:
        return None

    sqrt_disc = np.sqrt(disc)
    x_negative = (-b - sqrt_disc) / (2 * a)
    x_positive_bad = (-b + sqrt_disc) / (2 * a)
    x_positive_good = c / (a * x_negative)

    return x_negative, x_positive_bad, x_positive_good

#constants
a = 1
b = 1e8
c = 1

roots = quad_roots(a, b, c)

#identifies if real solutions exist or not
if roots is None:
    print("Real solutions does not exist.")
else:
    x_negative, x_positive_bad, x_positive_good = roots
    diff = abs(x_positive_bad - x_positive_good)
    rel_error = abs(diff / x_positive_good)
    print(f"(a) x-        = {x_negative}")
    print(f"(b) x+ (bad)  = {x_positive_bad}")
    print(f"(c) x+ (good) = {x_positive_good}")
    print(f"\nDifference between positive roots : {diff}")
    print(f"Error relative to good root    : {rel_error}")    
```

```python
#Part (d)
def rel_error_of_bad_form(b_values, dtype):
    #change the values of b to the desired floating-point precision
    b = b_values.astype(dtype)
    a = dtype(1.0)
    c = dtype(1.0)
    two = dtype(2.0)
    four = dtype(4.0)

    disc = b**2 - four * a * c
    sqrt_disc = np.sqrt(disc)

    x_negative = (-b - sqrt_disc) / (two * a)
    x_positive_bad = (-b + sqrt_disc) / (two * a)
    x_positive_good = c / (a * x_negative)

    #this will return the relative error of the "bad" formula
    return np.abs((x_positive_bad - x_positive_good) / x_positive_good)

b_values = np.logspace(1, 9, 500) #generating values of b from 10^4 to 10^9
error_float32 = rel_error_of_bad_form(b_values, np.float32)
error_float64 = rel_error_of_bad_form(b_values, np.float64)

#code to plot the graph
plt.figure()
plt.loglog(b_values, error_float32, label="32-bit (float32)", linewidth=2)
plt.loglog(b_values, error_float64, label="64-bit (float64)", linewidth=2)
plt.xlabel("b")
plt.ylabel("Relative error of 'bad' formula")
plt.title("Precision loss in the 'bad' $x_+$ formula")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.savefig('relative_error-x.png')
plt.show()
```

```python
x_value = 1.2e-8
result = f(x_value)
print("x =", x_value)
print("f(x) =", result)
print("Limit expected =", 0.5)

#uses the identity 1-cos(x) = 2sin^2(x/2)
def f_identity(x):
    return (2*np.sin(x/2)**2) /x**2

result_identity = f_identity(x_value)
print("f(x) resulting from trig identity =", result_identity)
```

```python
x_values = np.logspace(-1, -8, 200)

#obtains the y values by calling the function using direct computation and trig identity
y_manual = f(x_values)
y_trig_identity = f_identity(x_values)

#graphs the figures
plt.figure(figsize=(8,5))

#plots a logarithmic scale on x as y remains linear
plt.semilogx(x_values, y_manual, label="Direct Computation")
plt.semilogx(x_values, y_trig_identity, label="Trig Identity")

#demonstrates the limit by adding a horizontal line across the axis
plt.axhline(0.5, linestyle="--", label="Limit = 0.5")


plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r'Comparison between direct computing and trig identity as $x \rightarrow 0$')
plt.grid(True)
plt.legend()
plt.savefig('limit_comparison.png')
plt.show()
```

```python
# (a) Extremely small values of x, or values approaching zero, 
# cause catastrophic cancellation to occur. This is because 
# the value of e^x becomes close to 1.

# (b) Resolving catastrophic cancellation requires the use of
# of NumPy's expm1(x) function, which is specifically designed
# for e^x - 1

import numpy as np
import matplotlib.pyplot as plt

#functions that use manual computation and NumPy's special function
def fmanual(x):
    return (np.exp(x) - 1) / x

def fnumpy(x):
    return (np.expm1(x)) / x

x = np.logspace(-16, 0, 500)

#call the function to obtain y values
first_test = fmanual(x)
second_test = fnumpy(x)

#graphs the figures
plt.figure()

#plots a logarithmic scale on x as y remains linear
plt.semilogx(x, first_test, label="Manual")
plt.semilogx(x, second_test, label="NumPy's Special Function")

#demonstrates the limit by adding a horizontal line across the axis
plt.axhline(1.0, linestyle="--", label="Analytical limit = 1")

#demonstrates the area of cancellation
plt.axvspan(1e-16, 1e-14, alpha=0.2, label="Region of Cancellation")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(r"Manual vs NumPy's Special Function of $f(x) =\frac{e^x-1}{x}$")
plt.legend()
plt.grid(True)
plt.savefig('manual&numpy.png')
plt.legend()
plt.show()
```
