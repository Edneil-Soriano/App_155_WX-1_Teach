### Problem 1: Quadratic Equation (Gezerlis Problem 2.6)

Take the standard quadratic equation: $ax^2 + bx + c = 0$, whose solutions are 

$$x_{\pm}=\frac{-b \pm \sqrt{b^2-4ac}}{2a}$$

Take $b>0$ for concreteness. It is easy to see that when $b^2 >> ac$, we don't get a catastrophic cancellation when evaluating $b^2 - 4ac$ (we may still get a "benign" cancellation). Furthermore, $\sqrt{b^2-4ac} \approx b$. 

However, this means that $x_+$ will involve catastrophic cancellation in the numerator. We will employ an analytical trick in order to help us preserve significant figures. Observe that the product of the two roots must obey 

$$x_+ x_- = \frac{c}{a}$$

The answer now presents itself: use the standard quadratic formula to calculate $x_-$, for which no catastrophic cancellation takes place. Then, use $x_+ x_- = \frac{c}{a}$ to calculate $x_+$. Notice that you end up calculating $x_+$, via division only (i.e. without a catastrophic cancellation). 

Write a Python code (use `np.sqrt()`) that evaluates and prints out **(a)** $x_-$, **(b)** $x_+$ using the "bad formula", and **(c)** $x_+$ using the "good formula". 

Take $a=1$, $c=1$, $b=10^8$. Discuss the answers.

Lastly, **(d)** [FIGURE] Visualize the loss of precision by plotting the relative error of the "bad" $x_+$ formula. Vary $b$ using a log scaling (e.g., from $10^4$ to $10^9$). To highlight the effect of hardware limits, evaluate the bad formula using both 32-bit (np.float32) and 64-bit (np.float64) precision, comparing them against the "good" formula as the baseline.

```python
# a to c
import numpy as np
import matplotlib.pyplot as plt
import math

#constants
a, b, c = 1.0, 10**8, 1.0

discriminant = np.sqrt(b**2 - 4*a*c)

# a. x_- root
x_minus = (-b - discriminant) / (2*a)
# b. x_+ (bad formula)
x_plus_bad = (-b + discriminant) / (2*a)
# c. x_+ (good formula)
x_plus_good = (c / a) / x_minus


# print results
print("x_- =", x_minus)
print("x_+ (bad)  =", x_plus_bad)
print("x_+ (good) =", x_plus_good)
```

```python
# d [plot]

# define
quad = np.logspace(0.5, 9, 1000)

# Plug in quad = b
x_minus = ( - quad - np.sqrt(quad**2 - 4*a*c))/(2*a)

x_plus_bad = ( - quad + np.sqrt(quad**2 - 4*a*c))/(2*a)

x_plus_good = c/(a*x_minus)

# define the relative error
relError = np.abs((x_plus_bad - x_plus_good)/x_plus_good)

fig = plt.figure()
ax = fig.add_subplot(2,1,1)

#plot
plt.plot(quad,relError, 'r-', label = "Relative Error")
plt.yscale('log')
plt.xscale('log')
plt.savefig("Problem1.png")
```

#### Problem 1 code summary

The plot for Problem 1 visualizes the relative error between a standard quadratic formula and a stabilized alternative as a coefficient increases. Displayed on a log-log scale, the red line demonstrates that the standard formula suffers from a severe loss of precision. As the input value grows from $10^1$ up to $10^9$, the relative error climbs steadily from approximately $10^{-15}$ until it reaches nearly $10^0$. This steady upward trajectory on the logarithmic axes confirms that subtracting nearly equal large numbers causes catastrophic cancellation, rendering the direct numerical approach highly inaccurate for large values.

```python
import numpy as np
import matplotlib.pyplot as plt

# define the function
def f(x):
    return (1 - np.cos(x)) / x**2

# Generate x values from 0.1 to 10 with a step size of 0.1
x = np.arange(0.1, 10.1, 0.1)
y = f(x)

# Plot the results
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Plot of f(x) = (1 - cos(x)) / x^2')
plt.grid(True)
plt.savefig("Problem2_1.png")
plt.show()
```

```python
# b. As x -> 0, the limit (via L'Hopital's rule) of f(x) is equal to 1/2.

x_small = 1.2e-8
val = f(x_small)
print(f"f({x_small}) = {val}")
```

**(d)** Use a trigonometric identity to avoid the cancellation. Evaluate the new function at $x=1.2 \times 10^{-8}$ and compare with your analytical answer for $x \longrightarrow 0$.

**(e)** [FIGURE] Produce a figure that compares fow $f(x)$ approaches a certain value as $x \longrightarrow 0$, between directly computing $f(x)$ and by using a trigonometric identity to "avoid the cancellation". 

```python
def f_stable(x):
    return 2 * (np.sin(x / 2))**2 / x**2

val_stable = f_stable(x_small)
print(f"Stable f({x_small}) = {val_stable}")

# Generate exponentially small x values (e.g., from 10^-8 to 10^-1)
x_vals = np.logspace(-8, -1, 100)

y_direct = f(x_vals)
y_stable = f_stable(x_vals)

plt.figure(figsize=(8, 6))

# Plot both methods
plt.plot(x_vals, y_direct, label='Direct Computation (Unstable)', marker='o', markersize=3)
plt.plot(x_vals, y_stable, label='Trig Identity (Stable)', linestyle='--', linewidth=2)
plt.axhline(0.5, color='red', linestyle=':', label='Analytical Limit (0.5)')

# Use log scale for the x-axis to clearly see behavior near zero
plt.xscale('log')
plt.xlabel('x (log scale)')
plt.ylabel('f(x)')
plt.title('Comparison of Computations as x -> 0')
plt.legend()
plt.savefig("Problem2_2.png")
plt.grid(True)
plt.show()
```

```python
# a. Catastrophic cancellation occurs when x = 0.
import numpy as np
import matplotlib.pyplot as plt

#b. We can resolve the catastrophic cancelletion in f(x) by using the np.expm1 function, which computes e^x - 1 accurately without direct subtraction.



def f_manual(x): 
        return (np.exp(x) -1)/x

def f_special(x):
# np.expm1(x) handles the -1 internally to avoid cancellation
    return np.expm1(x) / x

x = np.logspace(-16, 0, 400)

y_man = f_manual(x)
y_sp = f_special(x)

plt.figure()
plt.semilogx(x, y_man, label = "Manual Calc")
plt.semilogx(x, y_sp, label = "Numpy special func")

#cancellation region in red
plt.axhline(1.0, linestyle="--", label="Cancellation region in f(x)")
plt.axvspan(1e-16, 1e-14, color='red', alpha=0.3, label="Cancellation Region")

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title("Manual vs Numpy special function")
plt.grid(True)
plt.legend()
plt.savefig("Problem2_3.png")
plt.show()

```

```python
import copy
from copy import deepcopy

original = [1, 2, [3, 4]] # notice the list inside a list
shallow_copied = original[:]

# modifying a top-level element
shallow_copied[0] = 99
print("After modifying top-level element:")
print(f"Original: {original}")        
print(f"Copied:   {shallow_copied}")  

# modifying a nested mutable element
shallow_copied[2][0] = "changed" 
print("\nAfter modifying nested element (SHALLOW):") # BTW \n in string means paragraph break
print(f"Original: {original}")        
print(f"Copied:   {shallow_copied}")  

deep_copied = deepcopy(original)
original = [1, 2, [3, 4]]
deep_copied[2][0] = "changed" 
print("\nAfter modifying nested element (DEEP COPIED):") # BTW \n in string means paragraph break
print(f"Original: {original}")        
print(f"Copied:   {deep_copied}")  
```
