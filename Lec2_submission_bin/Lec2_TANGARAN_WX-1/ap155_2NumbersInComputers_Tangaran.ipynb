### Problem 1: Quadratic Equation (Gezerlis Problem 2.6)

Take the standard quadratic equation: $ax^2 + bx + c = 0$, whose solutions are 

$$x_{\pm}=\frac{-b \pm \sqrt{b^2-4ac}}{2a}$$

Take $b>0$ for concreteness. It is easy to see that when $b^2 >> ac$, we don't get a catastrophic cancellation when evaluating $b^2 - 4ac$ (we may still get a "benign" cancellation). Furthermore, $\sqrt{b^2-4ac} \approx b$. 

However, this means that $x_+$ will involve catastrophic cancellation in the numerator. We will employ an analytical trick in order to help us preserve significant figures. Observe that the product of the two roots must obey 

$$x_+ x_- = \frac{c}{a}$$

The answer now presents itself: use the standard quadratic formula to calculate $x_-$, for which no catastrophic cancellation takes place. Then, use $x_+ x_- = \frac{c}{a}$ to calculate $x_+$. Notice that you end up calculating $x_+$, via division only (i.e. without a catastrophic cancellation). 

Write a Python code (use `np.sqrt()`) that evaluates and prints out **(a)** $x_-$, **(b)** $x_+$ using the "bad formula", and **(c)** $x_+$ using the "good formula". 

Take $a=1$, $c=1$, $b=10^8$. Discuss the answers.

Lastly, **(d)** [FIGURE] Visualize the loss of precision by plotting the relative error of the "bad" $x_+$ formula. Vary $b$ using a log scaling (e.g., from $10^4$ to $10^9$). To highlight the effect of hardware limits, evaluate the bad formula using both 32-bit (np.float32) and 64-bit (np.float64) precision, comparing them against the "good" formula as the baseline.

```python
import numpy as np
import matplotlib.pyplot as plt
 
# Part (a)-(c): "bad" vs "good" formula for x_+, at a fixed b
def quadratic_roots(a, b, c):
    disc = b**2 - 4 * a * c
    # Check if the quadratic equation has real solutions
    if disc < 0:
        return None
 
    sqrt_disc = np.sqrt(disc)
    x_minus = (-b - sqrt_disc) / (2 * a)
    x_plus_bad = (-b + sqrt_disc) / (2 * a)
    x_plus_good = c / (a * x_minus)
 
    return x_minus, x_plus_bad, x_plus_good
 
#Constants
a = 1
b = 1e8
c = 1

roots = quadratic_roots(a, b, c)

# Check whether real solutions exist
if roots is None:
    print("No real solutions.")
else:
    x_minus, x_plus_bad, x_plus_good = roots
    diff = abs(x_plus_bad - x_plus_good)
    rel_error = abs(diff / x_plus_good)
    print(f"(a) x-        = {x_minus}")
    print(f"(b) x+ (bad)  = {x_plus_bad}")
    print(f"(c) x+ (good) = {x_plus_good}")
    print(f"\nDifference between plus-roots : {diff}")
    print(f"Error relative to good root    : {rel_error}")
```

```python

# Part (d): relative error of the "bad" x_+ formula vs. b, for float32 and float64
def bad_formula_relative_error(b_values, dtype):
    # Convert b values to the desired floating-point precision
    b = b_values.astype(dtype)
    a = dtype(1.0)
    c = dtype(1.0)
    two = dtype(2.0)
    four = dtype(4.0)

    disc = b**2 - four * a * c
    sqrt_disc = np.sqrt(disc)

    x_minus = (-b - sqrt_disc) / (two * a)
    x_plus_bad = (-b + sqrt_disc) / (two * a)
    x_plus_good = c / (a * x_minus)

    # Return the relative error of the "bad" formula
    return np.abs((x_plus_bad - x_plus_good) / x_plus_good)


b_values = np.logspace(1, 9, 500)  # Generate b values from 10^4 to 10^9 
error_float32 = bad_formula_relative_error(b_values, np.float32)
error_float64 = bad_formula_relative_error(b_values, np.float64)

#plotting the graph
plt.figure()
plt.loglog(b_values, error_float32, label="32-bit (float32)", linewidth=2)
plt.loglog(b_values, error_float64, label="64-bit (float64)", linewidth=2)
plt.xlabel("b")
plt.ylabel("Relative error of 'bad' formula")
plt.title("Loss of precision in the 'bad' $x_+$ formula")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.savefig('rel_error-x.png')
plt.show()
```

**Bonus**: also compare part with different hardware floating-point precisions: 16/32/64/128-bit. 

```python
def f_multi(x_values, dtype, use_identity):
    """
    Evaluate f(x) = (1 - cos x) / x^2 at a given floating-point precision.
    """
    x = x_values.astype(dtype)

    if use_identity:
        # Use 1-cos(x) = 2 sin^2(x/2) identity
        two = dtype(2.0)
        return two * np.sin(x / two)**2 / x**2
    else:
        # Direct computation
        one = dtype(1.0)
        return (one - np.cos(x)) / x**2


x_values = np.logspace(-1, -8, 200)

#different float precisions to use
precisions = [
    ("16-bit", np.float16),
    ("32-bit", np.float32),
    ("64-bit", np.float64),
    ("128-bit", np.float128),
]

# Create a 2x2 grid of subplots with shared x- and y-axes.
fig, axes = plt.subplots(2, 2, figsize=(11, 8), sharex=True, sharey=True)

# Ignore warnings caused by underflow and division by zero at very small x.
with np.errstate(divide="ignore", invalid="ignore"):
    for ax, (label, dtype) in zip(axes.flat, precisions):
        
        # Compute f(x) using the direct form and the trig identity.
        y_dir = f_multi(x_values, dtype, use_identity=False)
        y_id = f_multi(x_values, dtype, use_identity=True)

        # Plot both methods using a logarithmic x-axis.
        ax.semilogx(x_values, y_dir, label="Direct")
        ax.semilogx(x_values, y_id, label="Trig identity")

        # Show the expected limit of f(x) as x approaches 0.
        ax.axhline(0.5, linestyle="--", color="k", linewidth=1, label="Limit = 0.5")

        # Add labels, title, and grid to each subplot.
        ax.set_title(label)
        ax.set_xlabel("x")
        ax.set_ylabel("f(x)")
        ax.grid(True, which="both", alpha=0.4)

# Display the legend in the first subplot.
axes.flat[0].legend()

fig.suptitle(r"Direct vs. trig-identity $f(x)$ as $x \rightarrow 0$, by precision")
fig.tight_layout()
plt.show()
```

```python
#a. Catastrophic cancellation occurs for very small x or as it approaches 0, because e^x becomes very close to 1.

#b. This can be resolved through numpy's expm1(x) function, designed for e^x - 1 and avoids catastrophic cancellation.

# code for item c

import numpy as np
import matplotlib.pyplot as plt

#functions to compute manually and with numpy's special function
def f_manual(x):
    return(np.exp(x) - 1) / x

def f_numpy(x):
    return (np.expm1(x))/x

x = np.logspace(-16, 0, 500)

#getting y values by calling the functions
y_manual = f_manual(x)
y_numpy = f_numpy(x)

#graphing the figures
plt.figure()
#plt.semilogx as x is scaled logarithmically while y stays linear
plt.semilogx(x, y_manual, label="Manual")
plt.semilogx(x, y_numpy, label="Numpy Special Function")

#plt.axhline to see the limit
plt.axhline(1.0, linestyle="--", label="Limit = 1")

#plt.axvspan to see the area of cancellation region
plt.axvspan(1e-16, 1e-14, alpha=0.2, label="Cancellation region")

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r"$f(x)=\frac{e^x-1}{x}$: Manual vs Numpy Special Function")
plt.grid(True)
plt.legend()
plt.savefig('man-numpy.png')
plt.show()

```
