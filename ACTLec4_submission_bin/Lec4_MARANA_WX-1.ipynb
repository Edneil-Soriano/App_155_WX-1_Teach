### Problem 1: Finite Difference

Let $$f(x) = 1+\frac{1}{2}\tanh(2x)$$

**(a)** Create functions that take as input an array or list of points corresponding to a mathematical function and output the derivative using **(a.1)** forward-difference, **(a.2)** backward-difference, and **(a.3)** central-difference methods.


```python
import numpy as np

def f(x):
    return 1 + (np.tanh(2*x))/2

#a.1 forward difference
def forward_dif(f, x, h):
    fd = (f(x+h)- f(x))/h
    return fd

#a.2 backward difference
def backward_dif(f, x, h):
    bd = (f(x)- f(x-h))/h
    return bd

#a.3 central difference
def central_dif(f, x, h):
    cd = (f(x+h/2) - f(x-h/2))/h
    return cd
```

**(b)** Calculate an analytic formula for the derivative. Make three subplots for the derivative of $f'(x)$, one for each method and the analytic answer on the same plots. Plot the exact answer (use `numpy`) as lines and the numerical one as dots. Describe how the two results compare. Use `plt.subplots`

```python
import numpy as np
import matplotlib.pyplot as plt

def fprime(x):
    return 1/((np.cosh(2*x))**2)

x = np.linspace(-3, 3, 200)
h = 0.1

plt.figure(figsize=(15,4))

# plotting forward difference
plt.subplot(1,3,1)
plt.plot(x, fprime(x), label="Analytic")
plt.plot(x, forward_dif(f,x,h), '.', label="Numerical")
plt.title("Forward Difference")
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.grid()
plt.legend()

# plotting backward difference
plt.subplot(1,3,2)
plt.plot(x, fprime(x), label="Analytic")
plt.plot(x, backward_dif(f,x,h), '.', label="Numerical")
plt.title("Backward Difference")
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.grid()
plt.legend()

# plotting central difference
plt.subplot(1,3,3)
plt.plot(x, fprime(x), label="Analytic")
plt.plot(x, central_dif(f,x,h), '.', label="Numerical")
plt.title("Central Difference")
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.grid()
plt.savefig('fig1.png', format='png', dpi=300)
plt.legend()
```

### Problem 2: Simple Harmonic Oscillator

The position $x(t)$ of a 1D simple harmonic oscillator is described by the equation of motion:

$$-kx = \frac{d^2x}{dt^2}$$

We know a solution can take the form of $x(t) = \cos(t)$. Calculate the velocity of this particle using any finite-difference method, and plot its absolute error with respect to the analytical value by varying your step-sizes. The absolute error should have the same trend as the $\mathcal{O}$ error of your chosen method.

```python
import math
import matplotlib.pyplot as plt

# position function
def x(t):
    return math.cos(t)

# analytical velocity: dx/dt= -sin(t)
def analytical_v(t):
    return -math.sin(t)

# using central difference for velocity
def central_dif(f,t,h):
    return (f(t+h/2) - f(t-h/2))/h

# using t0 as some fixed point to test
t0 = 10.0

hs = np.linspace(0.01, 2, 100)
errors_central = []
for h in hs:
    numerical_v = central_dif(x, t0, h)
    error = abs(numerical_v - analytical_v(t0))
    errors_central.append(error)

errors_backwards = []
for h in hs:
    numerical_v = backward_dif(x, t0, h)
    error = abs(numerical_v - analytical_v(t0))
    errors_backwards.append(error)

errors_forward = []
for h in hs:
    numerical_v = forward_dif(x, t0, h)
    error = abs(numerical_v - analytical_v(t0))
    errors_forward.append(error)

# plot absolute error vs step size

plt.plot(hs, errors_central, label="Central Difference")
plt.plot(hs, errors_backwards, label="Backward Difference")
plt.plot(hs, errors_forward, label="Forward Difference")
plt.xlabel("Step Size (h)")
plt.ylabel("Absolute Error")
plt.title("Central Difference Absolute Error")
plt.legend()
plt.grid()
plt.savefig('fig6_10.png', format='png', dpi=300) 
plt.show()
```

#### Problem 2 code summary

The code above tests all three finite difference methods. The velocity of a simple harmonic oscillator with position $x(t)$ = $cos(t)$ using central-difference, backward difference, and forward difference then compares it to the known analytical velocity $-sin(t)$ across a range of step sizes h.

### Derivatives

The derivate of a function $f(x) $ can be approximated using the **forward difference** as

$$f'(x) = \frac{f(x+h) - f(x)}{h} + \mathcal{O}(h)$$

using the **backward difference**

$$f'(x) = \frac{f(x) - f(x-h)}{h} + \mathcal{O}(h)$$

and the **central difference**

$$f'(x) = \frac{f(x + \frac{h}{2}) - f(x - \frac{h}{2})}{h} + \mathcal{O}(h^2)$$

where $h$ is a small step-size. To minimzie error, you can use Richardson Extrapolation, which can be found on the book (Gezerlis).

```python
import matplotlib.pyplot as plt

def plotex(cxs,cys,dxs,dys):
    plt.xlabel('x', fontsize=20)
    plt.ylabel('f(x)', fontsize=20)
    plt.plot(cxs, cys, 'r-', label='one function')
    plt.plot(dxs, dys, 'b--^', label='other function')
    plt.legend()
    plt.show()

cxs = [0.1*i for i in range(60)]
cys = [x**2 for x in cxs]
dxs = [i for i in range(7)]
dys = [x**1.8 - 0.5 for x in dxs]

plotex(cxs, cys, dxs, dys)
```
