### Problem 1: Finite Difference

Let $$f(x) = 1+\frac{1}{2}\tanh(2x)$$

**(a)** Create functions that take as input an array or list of points corresponding to a mathematical function and output the derivative using **(a.1)** forward-difference, **(a.2)** backward-difference, and **(a.3)** central-difference methods.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
h = 0.01
x = np.arange(-2,2,h)
y = 1 + 1/2 * np.tanh(2*x)

def forward_diff(x=x,y=y,h=h):
    fy = 1 + 1/2 * np.tanh(2*(x + h))
    dx = h
    dy = fy - y
    dy_dx = dy / dx
    return dy_dx

def backward_diff(x=x,y=y,h=h):
    by = 1 + 1/2 * np.tanh(2*(x - h))
    dx = h
    dy = y - by
    dy_dx = dy / dx
    return dy_dx

def central_diff(x=x,y=y,h=h):
    cfy = 1 + 1/2 * np.tanh(2*(x + h/2))
    cby = 1 + 1/2 * np.tanh(2*(x - h/2))
    dx = h
    dy = cfy - cby
    dy_dx = dy / dx
    return dy_dx
```

Plot the exact answer (use `numpy`) as lines and the numerical one as dots. Describe how the two results compare. Use `plt.subplots`

```python
h = 0.01
x = np.arange(-2,2,h)
y = 1 + 1/2 * np.tanh(2*x)
exact_dy_dx = np.gradient(y,h)


fig, ax = plt.subplots(1,3,figsize=(15,5))
fig.suptitle(r"Derivative of $f(x) = 1 + \frac{1}{2} \tanh{(x)}$", fontsize=18)

def forward_diff(x=x,y=y,h=h):
    fy = 1 + 1/2 * np.tanh(2*(x + h))
    dx = h
    dy = fy - y
    dy_dx = dy / dx
    ax[0].scatter(x,dy_dx,color='r',label='Numerical Plot',s=3)
    ax[0].plot(x,exact_dy_dx,color='k',label='Exact Plot')
    ax[0].set_title("Forward-Difference")
    ax[0].grid(True)
    ax[0].legend()
    ax[0].set_xlabel("$x$")
    ax[0].set_ylabel("$f'(x)$")

def backward_diff(x=x,y=y,h=h):
    by = 1 + 1/2 * np.tanh(2*(x - h))
    dx = h
    dy = y - by
    dy_dx = dy / dx
    ax[1].scatter(x,dy_dx,color='r',label='Numerical Plot',s=3)
    ax[1].plot(x,exact_dy_dx,color='k',label='Exact Plot')
    ax[1].set_title("Backward-Difference")
    ax[1].grid(True)
    ax[1].legend()
    ax[1].set_xlabel("$x$")

def central_diff(x=x,y=y,h=h):
    cfy = 1 + 1/2 * np.tanh(2*(x + h/2))
    cby = 1 + 1/2 * np.tanh(2*(x - h/2))
    dx = h
    dy = cfy - cby
    dy_dx = dy / dx
    ax[2].scatter(x,dy_dx,color='r',label='Numerical Plot',s=3)
    ax[2].plot(x,exact_dy_dx,color='k',label='Exact Plot')
    ax[2].set_title("Central-Difference")
    ax[2].grid(True)
    ax[2].legend()
    ax[2].set_xlabel("$x$")

forward_diff()
backward_diff()
central_diff()

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

### Problem 2: Simple Harmonic Oscillator

The position $x(t)$ of a 1D simple harmonic oscillator is described by the equation of motion:

$$-kx = \frac{d^2x}{dt^2}$$

We know a solution can take the form of $x(t) = \cos(t)$. Calculate the velocity of this particle using any finite-difference method, and plot its absolute error with respect to the analytical value by varying your step-sizes. The absolute error should have the same trend as the $\mathcal{O}$ error of your chosen method.

```python
import numpy as np
import matplotlib.pyplot as plt

n = 20
h = np.arange(0.001,3.0,0.001)
k = 2
t = 1.67

true_vel = -k*np.sin(k*t)

def fact(n):
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2,n+1):
        result *= (i)
    return result

coeffs = []
for i in range(n):
    c = (-1)**(i) / fact(2*i)
    coeffs.append(c)

def pos(t=t,k=k,n=n):
    x = (k*t)**2
    val = 0
    for coefficient in reversed(coeffs):
        val = val * x + coefficient
    return val

def v_forward(step,t=t,k=k):
    dv = pos(t + step) - pos(t)
    dt = step
    vel = dv / dt
    return vel

def v_central(step,t=t,k=k):
    dv = pos(t + step/2) - pos(t - step/2)
    dt = step
    vel = dv / dt
    return vel

def v_backward(step,t=t,k=k):
    dv = pos(t) - pos(t - step)
    dt = step
    vel = dv / dt
    return vel

vforward = np.array([v_forward(step) for step in h])
forward_error = abs(true_vel - vforward)

vcentral = np.array([v_central(step) for step in h])
central_error = abs(true_vel - vcentral)

vbackward = np.array([v_backward(step) for step in h])
backward_error = abs(true_vel - vbackward)

plt.title(rf"Velocity of Particle at time $t={t}\,[s]$")
plt.xlabel("Step Size $h$")
plt.ylabel("Absolute Error")
plt.plot(h, forward_error, label = "Forward-Difference", ls='--', color='r')
plt.plot(h, central_error, label = "Central-Difference", ls='-', color='b')
plt.plot(h, backward_error, label = "Backward-Difference", ls=':', color='g')
plt.grid(True)
plt.legend()

plt.show()
```

#### Problem 2 code summary

In this problem, we define first the position function $x(t) = \cos{(kt)}$ using its Taylor Expansion. We then apply the finite-difference approximations to that Taylor Expansion to evaluate the velocity $v(t)=-k\sin{(kt)}$ at a given time $t=1.67\,[s]$. Finally, we compare these evaluations and their errors from the true velocity for varying step sizes $h \in (0,3.0)$.

### Derivatives

The derivate of a function $f(x) $ can be approximated using the **forward difference** as

$$f'(x) = \frac{f(x+h) - f(x)}{h} + \mathcal{O}(h)$$

using the **backward difference**

$$f'(x) = \frac{f(x) - f(x-h)}{h} + \mathcal{O}(h)$$

and the **central difference**

$$f'(x) = \frac{f(x + \frac{h}{2}) - f(x - \frac{h}{2})}{h} + \mathcal{O}(h^2)$$

where $h$ is a small step-size. To minimzie error, you can use Richardson Extrapolation, which can be found on the book (Gezerlis).
