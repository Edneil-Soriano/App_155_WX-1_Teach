```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#finite difference
def function_1(x):
    for i in x:
        y = 1 + 0.5*np.tanh(2*x)
    return y

def forward_derivative(x, y):
    z1 = np.insert(y, 0, 0)
    z2 = np.append(y, 0)
    n = z2-z1
    m = n[1:-1]
    m = m*len(x)/(max(x)-min(x))
    return m
   

def backward_derivative(x, y):
    z2 = np.insert(y, 0, 0)
    z1 = np.append(y, 0)
    n = z1-z2 
    m = n[1:-1]
    m = m*len(x)/(max(x)-min(x))
    return m


def central_derivative(x, function_1):
    y = function_1(x+0.5*((max(x)-min(x))/len(x)))
    y_ = function_1(x-0.5*((max(x)-min(x))/len(x)))
    n = y-y_ 
    m = n[1:-1]
    m = m*len(x)/(max(x)-min(x))
    return m
   

x = np.linspace(-10, 10, num=1000)
y = function_1(x)
z = backward_derivative(x, y)
x_ = np.linspace(-10, 10, num=len(z))
x__ = np.linspace(-10, 10, num=998)
w = forward_derivative(x, y)
v = central_derivative(x, function_1)

fig, ax1 = plt.subplots()

ax1.plot(x,y)
ax1.plot(x_,z)
ax1.plot(x_,w)
ax1.plot(x__, v)


```
