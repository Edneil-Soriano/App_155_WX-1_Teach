### Problem 1: Horner's Rule (Gezerlis Problem 2.11)

This problem discusses error buildup when trying to evaluate polynomials without and with the use of *Horner's rule*. Take a polynomial of degree $n - 1$:

$$P(x) = p_0 + p_1 x + p_2 x^2 + p_3 x^3 + \dots + p_{n-1} x^{n-1} \tag{2.98}$$

Write a function that takes in a list containing the $p_i$'s, say `coeffs`, and the point $x$ and evaluates the value of $P(x)$ in the naive way, i.e., from left to right like how you would normally do it. You are not allowed to import any library such as `numpy`.


```python
def poly(x=11.01,n=8):
    coeffs = [(-11)**i for i in reversed(range(n))]
    val = 0
    for c in range(n): 
        term = coeffs[c] * (x ** c)
        val += term
    return val

poly()
```

Notice that this way of coding up the polynomial contains several (needless) multiplications. This is so because $x^i$ is evaluated as $x \times x \times \dots \times x$ (where there are $i - 1$ multiplications). Thus, this way of approaching the problem corresponds to:

$$1 + 2 + \dots + n - 2 = \frac{(n - 1)(n - 2)}{2} \tag{2.99}$$

multiplications, from $x^2$ all the way up to $x^{n-1}$. If we rewrite the polynomial:

$$P(x) = p_0 + x(p_1 + x(p_2 + x(p_3 + \dots + x(p_{n-2} + xp_{n-1})\dots))) \tag{2.100}$$

then we can get away with only $n - 1$ multiplications (i.e., no powers are evaluated). This is obviously more efficient, but equally important is the fact that this way we can substantially limit the accumulation of rounding error (especially for polynomials of large degree). 

Write a function that takes in a list containing the $p_i$'s, say `coeffs`, and the point $x$ and evaluates the value of $P(x)$ in the new way, i.e., from right to left. You are not allowed to import any library such as `numpy`. Apply the previous two functions to the (admittedly artificial) case of:

`coeffs = [(-11)**i for i in reversed(range(8))]`  
and `x = 11.01`. Observe any discrepancy and discuss its origin.

```python
def horner(x=11.01, n=8):
    coeffs = [(-11)**i for i in reversed(range(n))]
    val = 0
    for c in reversed(coeffs):
        val = val * x + c
    return val

horner()
```

Extend your work by testing both functions over increasing polynomial degrees $n \in [1, 500]$ evaluated at $x=2$. Measure execution times with the `time` library and track the absolute deviation $|P_{\text{naive}}(x) - P_{\text{Horner}}(x)|$. Plot the results side by side with `matplotlib` or `seaborn`.

```python
import time
import numpy as np
import matplotlib.pyplot as plt

x = 2
degree = np.arange(1,501,2)

def poly(x,n):
    coeffs = [(-0.5)**i for i in reversed(range(n))]
    val = 0
    for c in range(n): 
        term = coeffs[c] * (x ** c)
        val += term
    return val

def horner(x, n):
    coeffs = [(-0.5)**i for i in reversed(range(n))]
    val = 0
    for c in reversed(coeffs):
        val = val * x + c
    return val


poly_vals = [poly(x,n) for n in degree]
horner_vals = [horner(x,n) for n in degree]
abserr = [abs((poly(x,n) - horner(x,n))) for n in degree]

fig, ax = plt.subplots(2,1, figsize=(6,8))
fig.suptitle("Evaluation of Polynomials using the Traditional Method vs. Horner's Rule")
ax[0].plot(degree, poly_vals, label="Naive Method")
ax[0].plot(degree, horner_vals, label="Horner's Rule")
ax[0].set_yscale('log')
ax[0].set_ylabel(r"$P(2)$")
ax[0].legend()
ax[0].grid()

ax[1].plot(degree, abserr, color='r')
ax[1].set_ylabel("Absolute Error")
ax[1].set_xlabel("Polynomial Degree $n-1$")
ax[1].grid()

plt.show()
```

```python
time_poly = []
for n in degree:
    start_time = time.perf_counter()
    poly(x,n)
    end_time = time.perf_counter()
    exe_time = end_time - start_time
    time_poly.append(exe_time)

plt.plot(degree, time_poly, label='Naive Function')

time_horner = []
for n in degree:
    start_time = time.perf_counter()
    horner(x,n)
    end_time = time.perf_counter()
    exe_time = end_time - start_time
    time_horner.append(exe_time)

plt.plot(degree, time_horner, label="Horner's Rule")
plt.title("Execution Time using Naive Method vs. Horner's Rule")
plt.xlabel("Polynomial Degree $n-1$")
plt.ylabel(r"Execution Time")
plt.legend()
plt.grid(True)
plt.show()
```

### Problem 2: Taylor Series Evaluation

By applying Horner's rule without using any libraries, create a function to evaluate $f(x) = e^x$ based on its Taylor series approximation. Plot for $x \in [0,10]$. Compare with `np.exp()`.

```python
import numpy as np
import matplotlib.pyplot as plt

n = 100

coeffs = [1.0]
c = 1.0
for i in range(n):
    c /= (1+i)
    coeffs.append(c)

def exp_horner(x, n=n):
    val = 0
    for coefficient in reversed(coeffs):
        val = val * x + coefficient
    return val

domain = np.linspace(0,10,101)
value_exphorner = [exp_horner(x) for x in domain]
value_expnumpy = [np.exp(x) for x in domain]
abserr = [abs(np.exp(x) - exp_horner(x)) for x in domain]

plt.plot(domain, value_exphorner, label = f'Taylor Expansion to $n = {n}$', color='b', lw=3)
plt.plot(domain, value_expnumpy, label = 'NumPy Function', color='g')
plt.plot(domain, abserr, label = 'Absolute Error', color='r')
plt.title('Exponential Function $e^x$')
plt.xlabel('$x$')
plt.ylabel('$e^x$')
plt.grid(True)
plt.legend()

plt.show()
```

Again, by applying Horner's rule without using any libraries, create function to evaluate $f(x) = \sin(x)$ using its Taylor series approximation. Can you avoid evaluating exponents completely? If not, how can you minimize these operations? Plot the result for $x \in [0,10]$. Compare with `np.sin()`.

```python
n = 100

def fact(n):
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2,n+1):
        result *= (i)
    return result

coeffs = []
for i in range(n):
    c = (-1)**(i) / fact(1+(2*i))
    coeffs.append(c)

def sin_horner(x, n=n):
    y = x**2
    val = 0
    for coefficient in reversed(coeffs):
        val = val * y + coefficient
    return x*val

domain = np.linspace(0,10,101)
value_sinhorner = [sin_horner(x) for x in domain]
value_sinnumpy = [np.sin(x) for x in domain]
abserr = [abs(np.sin(x) - sin_horner(x)) for x in domain]

plt.plot(domain, value_sinhorner, label = f'Taylor Expansion to $n = {n}$', color='b',lw=3)
plt.plot(domain, value_sinnumpy, label = 'NumPy Function', color='g')
plt.plot(domain, abserr, label = 'Absolute Error', color='r')
plt.title(r'Sine Function $\sin{(x)}$')
plt.xlabel('$x$')
plt.ylabel(r'$\sin{(x)}$')
plt.grid(True)
plt.legend()

plt.show()
```

#### Problem 2 code summary

Here, we explore the use of Horner's Rule in Taylor Expansion and then comparing it with the numpy functions. The Taylor Expansion essentially gives us a polynomial with infinite terms (for practical purpose, we truncate the Taylor Expansion and evaluate it only up to $n = 100$). The first part is quite straightforward as we did not omit odd/even powers; for the second part, however, we define the factorial separately to cleanly evaluate the coefficients.

```python
import copy
from copy import deepcopy

original = [1, 2, [3, 4]] # notice the list inside a list
shallow_copied = original[:]

# modifying a top-level element
shallow_copied[0] = 99
print("After modifying top-level element:")
print(f"Original: {original}")        
print(f"Copied:   {shallow_copied}")  

# modifying a nested mutable element
shallow_copied[2][0] = "changed" 
print("\nAfter modifying nested element (SHALLOW):") # BTW \n in string means paragraph break
print(f"Original: {original}")        
print(f"Copied:   {shallow_copied}")  

deep_copied = deepcopy(original)
original = [1, 2, [3, 4]]
deep_copied[2][0] = "changed" 
print("\nAfter modifying nested element (DEEP COPIED):") # BTW \n in string means paragraph break
print(f"Original: {original}")        
print(f"Copied:   {deep_copied}")  
```
