### Problem 1: Horner's Rule (Gezerlis Problem 2.11)

This problem discusses error buildup when trying to evaluate polynomials without and with the use of *Horner's rule*. Take a polynomial of degree $n - 1$:

$$P(x) = p_0 + p_1 x + p_2 x^2 + p_3 x^3 + \dots + p_{n-1} x^{n-1} \tag{2.98}$$

Write a function that takes in a list containing the $p_i$'s, say `coeffs`, and the point $x$ and evaluates the value of $P(x)$ in the naive way, i.e., from left to right like how you would normally do it. You are not allowed to import any library such as `numpy`.


```python
def list_coeffs(coeffs, x):
    total = 0
    n = len(coeffs)
    for i, p_i in enumerate(coeffs):
        # power is number of coefficients minus 1, then the power of x is reduced by one very time it moves on to a new coeff
        power = n - 1 - i
        total += p_i * (x**power)
        # p_i is the coefficient at that specific index while i is the power of x
        # x basically adjusts to how many coefficients there are, meaning the highest exponent x will have is i - 1
    return total
```

```python
# testing the code

#list of the coefficients
poly_list = [1, 6, 87, 3]
# the value of x
x_val = 2

result = list_coeffs(poly_list, x_val)
print(result)
```

Notice that this way of coding up the polynomial contains several (needless) multiplications. This is so because $x^i$ is evaluated as $x \times x \times \dots \times x$ (where there are $i - 1$ multiplications). Thus, this way of approaching the problem corresponds to:

$$1 + 2 + \dots + n - 2 = \frac{(n - 1)(n - 2)}{2} \tag{2.99}$$

multiplications, from $x^2$ all the way up to $x^{n-1}$. If we rewrite the polynomial:

$$P(x) = p_0 + x(p_1 + x(p_2 + x(p_3 + \dots + x(p_{n-2} + xp_{n-1})\dots))) \tag{2.100}$$

then we can get away with only $n - 1$ multiplications (i.e., no powers are evaluated). This is obviously more efficient, but equally important is the fact that this way we can substantially limit the accumulation of rounding error (especially for polynomials of large degree). 

Write a function that takes in a list containing the $p_i$'s, say `coeffs`, and the point $x$ and evaluates the value of $P(x)$ in the new way, i.e., from right to left. You are not allowed to import any library such as `numpy`. Apply the previous two functions to the (admittedly artificial) case of:

`coeffs = [(-11)**i for i in reversed(range(8))]`  
and `x = 11.01`. Observe any discrepancy and discuss its origin.

```python
def horner_coeffs(coeffs, x):
    total = 0
    for p_i in coeffs:
        total = total * x + p_i
        
    return total
```

```python
# comparing both naive and horner solutions
coeffs = [(-11)**i for i in reversed(range(8))]
x = 11.01

naive_result = list_coeffs(coeffs, x)
horner_result = horner_coeffs(coeffs, x)

print("Naive Result:", naive_result)
print("Horner Result:", horner_result)
print("Discrepancy:", horner_result - naive_result)
```

Extend your work by testing both functions over increasing polynomial degrees $n \in [1, 500]$ evaluated at $x=2$. Measure execution times with the `time` library and track the absolute deviation $|P_{\text{naive}}(x) - P_{\text{Horner}}(x)|$. Plot the results side by side with `matplotlib` or `seaborn`.

```python
import time
import matplotlib.pyplot as plt
import random

# pasting functions again
def naive_coeffs(coeffs, x):
    total = 0
    n = len(coeffs)
    for i, p_i in enumerate(coeffs):
        # power is number of coefficients minus 1, then the power of x is reduced by one very time it moves on to a new coeff
        power = n - 1 - i
        total += p_i * (x**power)
        # p_i is the coefficient at that specific index while i is the power of x
        # x basically adjusts to how many coefficients there are, meaning the highest exponent x will have is i - 1
    return total

def horner_coeffs(coeffs, x):
    total = 0
    for p_i in coeffs: # reversed is just enumerate but in reverse
        total = total * x + p_i
        
    return total

# lists for storing data
degrees = list(range(1, 501))
naive_times = []
horner_times = []
deviations = []

# x value
# for some reason using a clean integer results in the absolute deivation plot to not present properly
x = 2.14

# generating random coefficients
# Loop through polynomial degrees from 1 to 500
for n in degrees:
    # (n + 1 coefficients total)
    # uniform to avoid choosing very close numbers ig
    coeffs = [random.uniform(-100.0, 100.0) for _ in range(n + 1)]

# time measurements
    
    # measure naive method time
    start_time = time.perf_counter() # starts timer
    naive_res = list_coeffs(coeffs, x)
    end_time = time.perf_counter() # ends timer
    naive_times.append(end_time - start_time)
    
    # measure horner method time
    start_time = time.perf_counter() # starts timer
    horner_res = horner_coeffs(coeffs, x)
    end_time = time.perf_counter() # ends timer
    horner_times.append(end_time - start_time)
    
    # calculate absolute deviation |P_naive(x) - P_horner(x)|
    abs_dev = abs(naive_res - horner_res)
    deviations.append(abs_dev)
    
plt.figure(figsize=(14, 5))

# execution time plot
plt.subplot(1, 2, 1)
plt.plot(degrees, naive_times, label="Naive Method", alpha=0.7)
plt.plot(degrees, horner_times, label="Horner's Method", alpha=0.7)
plt.title("Execution Time vs. Polynomial Degree")
plt.xlabel("Degree (n)")
plt.ylabel("Time (seconds)")
plt.legend()
plt.grid(True)

# absolute deviation
plt.subplot(1, 2, 2)
plt.plot(degrees, deviations, color="red", label="Absolute Deviation")
plt.title("Absolute Deviation vs. Polynomial Degree")
plt.xlabel("Degree (n)")
plt.ylabel("Absolute Error")
plt.yscale("log")  # Using log scale since deviations can span multiple magnitudes
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig('Figure_1.png', dpi=150)
plt.show()
plt.close()
```

### Problem 2: Taylor Series Evaluation

By applying Horner's rule without using any libraries, create a function to evaluate $f(x) = e^x$ based on its Taylor series approximation. Plot for $x \in [0,10]$. Compare with `np.exp()`.

Again, by applying Horner's rule without using any libraries, create function to evaluate $f(x) = \sin(x)$ using its Taylor series approximation. Can you avoid evaluating exponents completely? If not, how can you minimize these operations? Plot the result for $x \in [0,10]$. Compare with `np.sin()`.

```python
import numpy as np
import matplotlib.pyplot as plt

# Taylor Series evaluation for f(x) = e^x
def taylor_exp(x, terms=50):
    # nested form: 1 + x*(1 + x/2*(1 + x/3*(...)))
    val = 1.0
    for i in range(terms, 0, -1):
        val = 1.0 + (x / i) * val
    return val

# Taylor Series evaluation for f(x) = sin(x)
def taylor_sin(x, terms=25):
    # Nested form: x * (1 - x^2/(2*3) * (1 - x^2/(4*5) * (...)))
    x_sq = x * x  # compute x^2 only once to minimize exponent operations
    val = 1.0
    for i in range(terms, 0, -1):
        denom = (2 * i) * (2 * i + 1)
        val = 1.0 - (x_sq / denom) * val
    return x * val

# plots
# Create fine-grained inputs from 0 to 10 for accurate plotting
x_vals = np.linspace(0, 10, 500)

# Evaluate exponential function
y_exp_custom = [taylor_exp(x) for x in x_vals]
y_exp_numpy = np.exp(x_vals)

# Evaluate sine function
y_sin_custom = [taylor_sin(x) for x in x_vals]
y_sin_numpy = np.sin(x_vals)

# Plotting the results
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plot for e^x
ax1.plot(x_vals, y_exp_numpy, label='np.exp()', color='blue', linewidth=2)
ax1.plot(x_vals, y_exp_custom, label='Horners Method exp()', color='orange', linestyle='--', linewidth=2)
ax1.set_title('Comparison for $f(x) = e^x$')
ax1.set_xlabel('x')
ax1.set_ylabel('f(x)')
ax1.legend()
ax1.grid(True)

# Plot for sin(x)
ax2.plot(x_vals, y_sin_numpy, label='np.sin()', color='green', linewidth=2)
ax2.plot(x_vals, y_sin_custom, label='Horners Method sin()', color='red', linestyle='--', linewidth=2)
ax2.set_title('Comparison for $f(x) = \sin(x)$')
ax2.set_xlabel('x')
ax2.set_ylabel('f(x)')
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.savefig('Figure_2.png', dpi=150)
plt.show()
plt.close()
```
