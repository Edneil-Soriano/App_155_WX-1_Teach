```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# recreating Euler method solution for simple pendulum

## important constants

g = 9.81
l = 1

## initial conditions

theta_0 = 0.5
w_0 = 0

dt_ = 0.001

angle = np.zeros(10000)
omega = np.zeros(10000)

angle[0] = theta_0
omega[0] = w_0

for i in range(len(angle)-1):
  angle[i+1] = angle[i] + dt_*omega[i]
  omega[i+1] = omega[i] -(g/l)*dt_*np.sin(angle[i])

plt.plot(angle)
```

```python
#runge-kutta

## important constants

g = 9.81
l = 1

## initial conditions

theta_0 = 0.5
w_0 = 0
t_0 = 0

dt = 0.001

rk_angle = np.zeros(10000)
rk_omega = np.zeros(10000)
rk_time = np.linspace(start=0, stop = 99.99, num=10000)

rk_angle[0] = theta_0
rk_omega[0] = w_0

for i in range(len(rk_angle)-1):

  K_0 = (-g/l)*np.sin(rk_angle[i])*(rk_time[i])
  K_1 = (-g/l)*np.sin(rk_angle[i] + (dt/2)*K_0)*(rk_time[i] + dt/2)
  K_2 = (-g/l)*np.sin(rk_angle[i] + (dt/2)*K_1)*(rk_time[i] + dt/2)
  K_3 = (-g/l)*np.sin(rk_angle[i] + K_2*dt)*(rk_time[i] + dt)

  rk_angle[i+1] = rk_angle[i] + (dt/6) * (K_0 + 2*K_1 + 2*K_2 + K_3)

  k_0 = (-g/l)*np.sin(rk_angle[i])
  k_1 = (-g/l)*np.sin(rk_angle[i] + (dt/2)*k_0)
  k_2 = (-g/l)*np.sin(rk_angle[i] + (dt/2)*k_1)
  k_3 = (-g/l)*np.sin(rk_angle[i] + k_2*dt)

  rk_omega[i+1] = rk_omega[i] + (dt/6)*(k_0 + 2*k_1 + 2*k_2 + k_3)

plt.plot(rk_angle)
plt.axhline(0.5)
```
