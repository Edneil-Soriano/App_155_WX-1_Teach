### Problem 1: 2D Wave interference (Gezerlis Problem 1.15)

Assume that the combined eﬀect of two objects dropped in a water basin can be modelled by:
$$ W(\textbf{r}) = A \sin (k|\textbf{r}-\textbf{r}_0|) + A  \sin (k|\textbf{r}-\textbf{r}_1|). $$ You can think of this as an updated version of Eq. (1.3) in the book, where we are dealing with the linear addition of sinusoidal waves, instead of electric-field contributions. For simplicity, take $A=1$, $k=2\pi/0.3$, $r_0 = (−1, 0)$ , and $r_1 = (1, 0)$. Since we are now faced with a density plot, you should employ, NOT `streamplot()` but, `imshow()` from `Matplotlib`; in each of $x$ and $y$ your plot should extend from $−4$ to $+4$.

```python
# insert your code here
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
# value for A
A = float(input())
```

```python
# value for k
k = 2*np.pi/0.3
```

```python
# value for x boundaries
x_lower_bound = float(input("X Lower Bound "))
x_upper_bound = float(input("X Upper Bound "))
```

```python
# value for y boundaries
y_lower_bound = float(input("Y Lower Bound "))
y_upper_bound = float(input("Y Upper Bound "))
```

```python
# step size
step_size = 500
```

```python
# r coordinates
r_0x = float(input("r_0 x value: "))
r_0y = float(input("r_0 y value: "))
```

```python
# planning to make code for changable r amounts, r_0 -> r_n
r_1x = float(input("r_1 x value: "))
r_1y = float(input("r_1 y value: "))
```

```python
# grid, using linspace is more efficient than calculating many plot points
x = np.linspace(x_lower_bound, x_upper_bound, step_size)
y = np.linspace(y_lower_bound, y_upper_bound, step_size)
X, Y = np.meshgrid(x, y)

# distance from two sources (r^2) since both X points are -1 and + 1 away and the Y axes are on the same level hence no + or - for Y
r_0 = np.sqrt((X + (r_0x))**2 + (Y + (r_0y))**2)
r_1 = np.sqrt((X + (r_1x))**2 + (Y + (r_1y))**2)

# main equation
W = A*np.sin(k*r_0) + A*np.sin(k*r_1)

# old plot base
# plt.plot(x, y)
# plt.title("2D Wave Plot")
# plt.xlabel("x")
# plt.ylabel("f(x)")
# plt.grid(True)

# new plot using figure and imshow, i like twilight and managua
plt.figure(figsize=(7, 7))
plt.imshow(W, extent=[x_lower_bound , x_upper_bound, y_lower_bound, y_upper_bound], origin='lower', cmap='managua')

# do not forget axes labels
plt.title("2D Wave interference")
plt.colorbar(label='W(r)')
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

# saving and displaying
plt.savefig('2D_Wave_Interference_Plot.png', dpi=150)
plt.show()
plt.close()
```

c. Create a function `transpose_matrix(orig_matrix)` which returns the transpose of any matrix also in "list-of-lists" form. Do not assume the size of the matrix; figure it out within the function. Double-check that your implementations above work. Create a $3\times4$ random integer matrix, print it out and its transpose. The values and shape between the two should be consistent.

```python
def transpose_matrix(orig_matrix):
    # getting dimensions of orig_matrix
    num_rows = len(orig_matrix)
    num_cols = len(orig_matrix[0])

    # sets up a matrix of the size with rows and columns swapped
    transposed_matrix = [[0 for _ in range(num_rows)] for _ in range(num_cols)]

    # swapping elements
    for r in range(num_rows):
        for c in range(num_cols):
            transposed_matrix[c][r] = int(orig_matrix[r][c])
    return transposed_matrix

# Test
nrows = 3
ncols = 5

# using random_integer_matrix(nrows, ncols) code from part a. to generate random matrix
orig_matrix = random_integer_matrix(nrows, ncols)

# prints original matrix for comparison
print("Original Matrix =")
print(orig_matrix)

# prints transposed matrix
print("Transposed Matrix =")
# rearranges list elements to appear at the bottom instead of at the right side
for row in transpose_matrix(orig_matrix):
    print(row)
```

d. Create a function `matrix_multiply(left_matrix, right_matrix)` which performs the matrix multiplication. Do not assume the matrices follow a specific shape. Here's some tips and strategies:
* Determine how many rows and columns the left and right matrices have. Which numbers should match in order for us to have valid matrix multiplication?
* Iterating the elements of "list-of-lists" gives us the rows, but we need to iterate the columns for the right matrix.
* Use the `transpose_matrix` function, then iteratively apply `scalar_product`.

```python
def matrix_multiply(left_matrix, right_matrix):
    # defining rows and columns using the general name
    rows_left_matrix = len(left_matrix)
    cols_left_matrix = len(left_matrix[0])
    rows_right_matrix = len(right_matrix)
    cols_right_matrix = len(right_matrix[0])

    if cols_left_matrix != rows_right_matrix:
    # raise Value Error intentionally crashes the program, using that instead of print so that we know why the error appeared with a message
        raise ValueError("Multiplication not Possible: Dimension Mismatch")

    # sets up a matrix of the size with Rows of A and Columns of B
    result = [[0 for _ in range(cols_right_matrix)] for _ in range(rows_left_matrix)]

    # The Multiplication Part
    for i in range(rows_left_matrix):          # iterates over Rows of Left matrix
        for j in range(cols_right_matrix):     # iterates over Columns of Right matrix
            for k in range (cols_left_matrix): # since cols_left_matrix = rows_right_matrix this iterates over Columns of Left and Rows of Right
                result[i][j] += int(left_matrix[i][k] * right_matrix[k][j])
    return(result)
```

```python
# generating a left matrix
left_matrix = random_integer_matrix(left_nrows, left_ncols)

# presents original matrices
print("Matrix A")
print(left_matrix)

# generating right matrix
right_matrix = random_integer_matrix(right_nrows, right_ncols)

# presents original matrices
print("Matrix B")
print(right_matrix)

# Auto transposing code

# Case 1: Already compatible (e.g., 2x3 and 3x4)
if left_ncols == right_nrows:
    print("Product Matrix =")
    # Presents Product Matrix Top-Down
    for row in matrix_multiply(left_matrix, right_matrix):
        print(row)

# Case 2: Match by transposing Matrix B (e.g., 2x3 and 4x3)
elif left_ncols == right_ncols:
    right_matrix = transpose_matrix(right_matrix)
    product = matrix_multiply(left_matrix, right_matrix)
    print("Right Matrix Was Transposed")
    print("Transposed Matrix B")
    for row in right_matrix:
        print(row)
    print("Product Matrix =")
    # Presents Product Matrix Top-Down
    for row in matrix_multiply(left_matrix, right_matrix):
        print(row)

# Case 3: Match by transposing Matrix A (e.g., 3x2 and 3x4)
elif left_nrows == right_nrows:
    left_matrix = transpose_matrix(left_matrix)
    product = matrix_multiply(left_matrix, right_matrix)
    print("Left Matrix Was Transposed")
    print("Transposed Matrix A")
    # Presents Transposed Matrix Top-Down
    for row in left_matrix:
        print(row)
    print("Product Matrix =")
    # Presents Product Matrix Top-Down
    for row in matrix_multiply(left_matrix, right_matrix):
        print(row)

# Case 4: Match by transposing both (e.g., 3x2 and 4x3)
elif left_nrows == right_ncols:
    left_matrix = transpose_matrix(left_matrix)
    right_matrix = transpose_matrix(right_matrix)
    product = matrix_multiply(left_matrix, right_matrix)
    print("Both Matrices Were Transposed")
    print("Transposed Matrix A")
    # Presents Transposed Matrix Top-Down
    for row in left_matrix:
        print(row)
    print("Transposed Matrix B")
    # Presents Transposed Matrix Top-Down
    for row in right_matrix:
        print(row)
    print("Product Matrix =")
    # Presents Product Matrix Top-Down
    for row in matrix_multiply(left_matrix, right_matrix):
        print(row)

# Case 5: No Matches (e.g., 3x4 and 5x6)
else:
    print("Product Matrix Cannot Be Obtained: Dimensional Mismatch")
```

### Visualizing Electric Fields 

The electric field anywhere given a set of $n$ point charges is given by
$$ \mathbf{E}(\mathbf{r}) = \sum_{i=0}^{n-1} kq_i \frac{\mathbf{r} - \mathbf{r}_i}{|\mathbf{r} - \mathbf{r}_i|^3} $$

```python
import numpy as np
import matplotlib.pyplot as plt
from math import sqrt
from copy import deepcopy

def makefield(xs, ys):
    qtopos = {1: (-1,0), -1: (1,0)}
    n = len(xs)
    Exs = [[0. for k in range(n)] for j in range(n)]
    Eys = deepcopy(Exs)
    for j,x in enumerate(xs):
        for k,y in enumerate(ys):
            for q, pos in qtopos.items():
                posx, posy  = pos
                R = sqrt((x - posx)**2 + (y - posy)**2)
                Exs[k][j] += q*(x - posx)/R**3
                Eys[k][j] += q*(y - posy)/R**3
    return Exs, Eys

def plotfield(boxl,n):
    xs = [-boxl + i*2*boxl/(n-1) for i in range(n)]
    ys = xs[:]
    Exs, Eys = makefield(xs, ys)
    xs=np.array(xs); ys=np.array(ys)
    Exs=np.array(Exs); Eys=np.array(Eys)
    plt.streamplot(xs, ys, Exs, Eys, density=1.5, color='tab:brown')
    plt.xlabel('$x$')
    plt.ylabel('$y$')
    plt.axis('square')
    plt.show()
    
plotfield(2., 50)
```
