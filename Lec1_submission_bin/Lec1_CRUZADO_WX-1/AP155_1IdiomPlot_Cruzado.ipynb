### Problem 1: 2D Wave interference (Gezerlis Problem 1.15)

Assume that the combined eﬀect of two objects dropped in a water basin can be modelled by:
$$ W(\textbf{r}) = A \sin (k|\textbf{r}-\textbf{r}_0|) + A  \sin (k|\textbf{r}-\textbf{r}_1|). $$ You can think of this as an updated version of Eq. (1.3) in the book, where we are dealing with the linear addition of sinusoidal waves, instead of electric-field contributions. For simplicity, take $A=1$, $k=2\pi/0.3$, $r_0 = (−1, 0)$ , and $r_1 = (1, 0)$. Since we are now faced with a density plot, you should employ, NOT `streamplot()` but, `imshow()` from `Matplotlib`; in each of $x$ and $y$ your plot should extend from $−4$ to $+4$.

```python
import matplotlib.pyplot as plt
import numpy as np

# Defining the constants as given in the equation.
A = 1
k = 2*np.pi/0.3
r1 = np.array([-1,0])
r2 = np.array([1,0])

# Creating the gridspace for vector r from -4 to 4.
x = np.linspace(-4,4,101)
y = np.linspace(-4,4,101)
xv, yv = np.meshgrid(x,y)

# Calculating the distance between the vector r and vectors r_0 and r_1.
rr1 = np.sqrt((xv-r1[0])**2 + (yv-r1[1])**2)
rr2 = np.sqrt((xv-r2[0])**2 + (yv-r2[1])**2)

# Defining the equation as given in the problem.
W = A*np.sin(k * abs(rr1)) + A*np.sin(k * abs(rr2))

# Plot
plt.imshow(W, cmap='inferno', extent=[-4,4,-4,4])
plt.xlabel("Horizontal Distance")
plt.ylabel("Vertical Distance")
plt.title("2D Wave Interference of Two Disturbances")
plt.savefig('Fig1.png', dpi=300, bbox_inches='tight')

#The goal of this problem is to generate a plot showing the interference patterns of two objects dropped on a water basin.
#We use np.linspace to create the domain of x and y, meshgrid to generate this 2D space, and make multivariable function and operations convenient.
#We define our function W(r) using the meshgrid variables xv and yv, taking into account that r is a radial vector and xv, yv are expressed in the cartesian coordinate system.
```

a. Create a function `random_integer_matrix(nrows, ncols)` that generates a matrix in the form of "list-of-lists" whose elements are random integers from $1$ to $9$. Each "sub-list" inside the main list is a row in the matrix.

```python
def random_integer_matrix(nrows, mcols):
    A = []
    for n in range(nrows):
        row = []
        for m in range(mcols):
            row.append(random.randint(1,9)) #This creates a single row full of random integers and repeats it m times.
        A.append(row) #This inserts the rows into the blank list/matrix A.
    return A

#The structure of the for-loops outline how we usually locate the entries of a matrix.
#We look for the rows, and then the columns first. From there, we work backwards and insert our random integers.
```

b. Create a function `scalar_product(list1, list2)` where you multiply each element at the same index in the two lists and sum/accumulate the results.

```python
def scalar_product(A,B):
    a = [entry for row in A for entry in row] #This 'flattens' Matrix A into a single row.
    b = [entry for row in B for entry in row] #This 'flattens' Matrix B into a single row.
    result = sum(x*y for x,y in zip(a,b)) #zip(a,b) matches the flattened matrices a and b and sums the product of corresponding terms.
    return result

#a and b are essentially flattened matrices to make zipping more convenient.
#By zipping them, we correspondingly pair entries of a and b to perform the multiplication, just as how we do it with dot product.
```

c. Create a function `transpose_matrix(orig_matrix)` which returns the transpose of any matrix also in "list-of-lists" form. Do not assume the size of the matrix; figure it out within the function. Double-check that your implementations above work. Create a $3\times4$ random integer matrix, print it out and its transpose. The values and shape between the two should be consistent.

```python
def transpose(A):
    A_trans = []
    for col in range(len(A[0])):
        new_row = []
        for row in range(len(A)):
            #The reference [row][col] searches for the specific entry ij in A.
            new_row.append(A[row][col]) 
        #The order at which the for-loop for col and row was structured inverts the manner in which each entry is appended.
        A_trans.append(new_row)
    return A_trans

#Unlike how we did it in generating the random integer matrix, we reverse the processes.
#Here, we convert rows into columns and vice versa, and append them into a blank list to build the transposed matrix.
```

d. Create a function `matrix_multiply(left_matrix, right_matrix)` which performs the matrix multiplication. Do not assume the matrices follow a specific shape. Here's some tips and strategies:
* Determine how many rows and columns the left and right matrices have. Which numbers should match in order for us to have valid matrix multiplication?
* Iterating the elements of "list-of-lists" gives us the rows, but we need to iterate the columns for the right matrix.
* Use the `transpose_matrix` function, then iteratively apply `scalar_product`.

```python
def matrix_multiplication(A,B):
    if len(A[0]) != len(B): #Checks if the number of columns of A matches the number of rows of B.
        print("Matrix Multiplication is NOT possible due to dimensional mismatch.")
        return None

    else:
        print("Matrix Multiplication is possible.")

        #B is transposed to easily zip the rows of A and columns of B.
        #This makes it convenient to use the scalar_product function.
        B_trans = transpose(B) 
        
        C = []
        for row_a in A:
            new_row = []
            for col_b in B_trans:
                val = scalar_product([row_a],[col_b])
                new_row.append(val)
            C.append(new_row)

        return C

A = random_integer_matrix(2,2)
B = random_integer_matrix(2,1)
print("Matrix A :", A)
print("Matrix B: ", B)
matrix_multiplication(A,B)

#First, we check if the matrices satisfy the condition for matrix multiplication; columns of A = rows of B.
#We transpose B to easily perform scalar product using the user-defined function.
#Finally, we build our matrix C from the values of the scalar product operation.
```

### Visualizing Electric Fields 

The electric field anywhere given a set of $n$ point charges is given by
$$ \mathbf{E}(\mathbf{r}) = \sum_{i=0}^{n-1} kq_i \frac{\mathbf{r} - \mathbf{r}_i}{|\mathbf{r} - \mathbf{r}_i|^3} $$

```python
import numpy as np
import matplotlib.pyplot as plt
from math import sqrt
from copy import deepcopy

def makefield(xs, ys):
    qtopos = {1: (-1,0), -1: (1,0)}
    n = len(xs)
    Exs = [[0. for k in range(n)] for j in range(n)]
    Eys = deepcopy(Exs)
    for j,x in enumerate(xs):
        for k,y in enumerate(ys):
            for q, pos in qtopos.items():
                posx, posy  = pos
                R = sqrt((x - posx)**2 + (y - posy)**2)
                Exs[k][j] += q*(x - posx)/R**3
                Eys[k][j] += q*(y - posy)/R**3
    return Exs, Eys

def plotfield(boxl,n):
    xs = [-boxl + i*2*boxl/(n-1) for i in range(n)]
    ys = xs[:]
    Exs, Eys = makefield(xs, ys)
    xs=np.array(xs); ys=np.array(ys)
    Exs=np.array(Exs); Eys=np.array(Eys)
    plt.streamplot(xs, ys, Exs, Eys, density=1.5, color='tab:brown')
    plt.xlabel('$x$')
    plt.ylabel('$y$')
    plt.axis('square')
    plt.show()
    
plotfield(2., 50)
```
