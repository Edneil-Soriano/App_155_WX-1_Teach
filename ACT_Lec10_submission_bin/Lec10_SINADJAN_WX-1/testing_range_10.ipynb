```python
import matplotlib.pyplot as plt
import numpy as np
import scipy 
```

```python
pi = np.pi

xs = np.linspace(-10, 10, num =1000)
n = xs.size

timestep = xs[1]-xs[0]


# frequency array 

freq = np.fft.fftfreq(n, d=timestep)

#triangle

ys = scipy.signal.sawtooth(xs, pi**-1*0.5)
yt = scipy.signal.sawtooth(xs, 0.3)
yv = np.sin(3*xs)


# sinusoid

ya = np.sin(10*2*pi*xs)
yb = np.sin(4*2*pi*xs)



# y sum

ysum = ys + yt

# fourier transform

fft = np.fft.fft(ysum)



#plotting

fig, ((ax1), (ax2)) = plt.subplots(1,2)

ax1.plot(xs, ysum)
ax2.plot(freq, np.abs(fft))
ax2.set_xlim(-3,3)
```
