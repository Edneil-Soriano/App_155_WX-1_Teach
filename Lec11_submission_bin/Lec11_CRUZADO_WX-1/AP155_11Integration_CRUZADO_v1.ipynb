The following figures above involve four different integratino methods, namely the Monte Carlo Method, Trapezoidal Rule, Simpson's 1/3 Rule (Parabolic), and Simpson's 3/8 Rule (Cubic).

The Monte Carlo Method involves random sampling in which we randomly plot a point on the graph and evaluate whether or not it is inside or  outside the area bounded by the axes and the curve. By taking the probabilty of having a point inside the bounded area $P = \frac{\text{Points inside the Curve}}{\text{Number of Points N}}$ and multiplying it by the box area, we get an estimation of the shaded area.

The Trapezoidal Rule is a method similar to the Riemann summation in which we divide and create trapezoids under the curve. By taking the area of each trapezoid $A=\frac{B_1+B_2}{2} \cdot h$, we get a better estimation of what the Riemann summation can provide.

The Simpson's 1/3 Rule approximates the area under the curve by fitting a series of parabolas through three adjacent points, rather than using straight line segments like the Trapezoidal Rule. By applying a weighted sum of functino values, we achieve a higher-order accuracy for smooth curves.

On the other hand, Simpson's 3/8 Rule extends this polynomials approximation approach by fitting cubic polynomials through sets of four points across three subintervals. This method is more reliablewhen the number of subintervals is a multiple of three.

---
Monte Carlo simulations work by making use of a random sampling method, large sample size, and the Law of Large Numbers, where the ratio of random points falling beneath the curve to the total points inside the bounded area. This ratio acts as a probability that describes the chance of choosing a random point that lands inside the bounded area. While Monte Carli simulations are suitable in high-dimensional spaces, the deterministic Newton-Cotes methods explored in this exercise offer a more reliable approximation for low-dimensional, smooth, and well-behaved functions.

```python
def monte_carlo(func, a, b, y_min, y_max, N = N):
    x_pts = np.random.uniform(a,b,N)
    y_pts = np.random.uniform(y_min, y_max, N)

    fx = func(x_pts)

    pos_hits = (y_pts > 0) & (y_pts <= fx)
    neg_hits = (y_pts < 0) & (y_pts >= fx)

    inside_mask = pos_hits | neg_hits
    outside_mask = ~inside_mask

    box_area = (b-a) * (y_max - y_min)
    
    integral = box_area * (np.sum(pos_hits) - np.sum(neg_hits)) / N
    shaded_area = box_area * (np.sum(inside_mask)) / N
    
    return {
        'integral': integral,
        'shaded_area': shaded_area,
        'x_pts': x_pts,
        'y_pts': y_pts,
        'inside_mask': inside_mask,
        'outside_mask': outside_mask
    }
```

The function above generates $N$ rnadom uniform points within a bounding box that encloses the function within a specified interval. Here, we account for both positive and negative values, points inside and outside the curve to compute the integral value. We then estimate the integral by multiplying the box area by the ratio of inside hits over the total sample size $N$.

```python
def trapezoid(func, a, b, n, plot=True):
    h = (b - a) / (n - 1)
    x_pts= a + np.arange(n) * h
    fx = func(x_pts)
    
    cs = np.ones(n); cs[0] = 0.5; cs[-1] = 0.5
    contribs = cs*func(x_pts)
    integral = h*np.sum(contribs)

    y1, y2 = fx[:-1], fx[1:]
    same_sign = (y1 * y2) >= 0

    interval_areas = np.zeros(n-1)
    interval_areas[same_sign] = 0.5 * h * (np.abs(y1[same_sign]) + np.abs(y2[same_sign]))

    diff_sign = ~same_sign
    if np.any(diff_sign):
        interval_areas[diff_sign] = 0.5 * h * (y1[diff_sign] + y2[diff_sign])

    shaded_area = np.sum(interval_areas)

    if plot:
        x_dense = np.linspace(a, b, 500)
        plt.plot(x_dense, func(x_dense), 'k--', label='$f(x)$ True Curve', alpha=0.7)
        
        for i in range(n - 1):
            px = [x_pts[i], x_pts[i], x_pts[i+1], x_pts[i+1]]
            py = [0, fx[i], fx[i+1], 0]
            plt.fill(px, py, color='skyblue', edgecolor='navy', alpha=0.4, linewidth=1.2)
        
        plt.plot(x_pts, fx, 'ro-', markersize=4, label='Trapezoid Linear Interpolant')
        plt.axhline(0, color='black', linewidth=0.8)

        plt.title(f"Trapozoidal Method \n (Estimated Area = {shaded_area:0.4f})")
        plt.xlabel('x')
        plt.ylabel('y')
        plt.legend(loc='upper right')
        plt.grid(True, ls=':', alpha=0.4)

    return {
        'integral': integral,
        'shaded_area': shaded_area,
        'x_pts': x_pts,
        'fx': fx
    }
```

We discretize the function along the specified interval into $n$ equally spaced nodes separated by a step size $h$. We then estimate the integral using the formula for the area of the trapezoid that was simplified into a term by term array with weights. Here, we include the geometric shaded area by summing the absolute, individual trapezoid areas.

```python
def simpson_para(func, a, b, n, plot=True):
    h = (b - a) / (n - 1)
    x_pts= a + np.arange(n) * h
    fx = func(x_pts)
    
    cs = 2 * np.ones(n); cs[1::2] = 4; cs[0] = 1; cs[-1] = 1
    contribs = cs*func(x_pts)
    integral = (h / 3) * np.sum(contribs)

    num_panels = (n - 1) // 2
    shaded_area = 0.0

    if plot:
         x_dense = np.linspace(a, b, 500)
         plt.plot(x_dense, func(x_dense), 'k--', label='$f(x)$ True Curve', alpha=0.7)
        
    for i in range(num_panels):
        idx = [2 * i, 2 * i + 1, 2 * i + 2]

        x_pan = x_pts[idx]
        y_pan = fx[idx]

        poly = np.polyfit(x_pan, y_pan, 2)

        x_sub = np.linspace(x_pan[0], x_pan[-1], 200)
        y_sub = np.polyval(poly, x_sub)

        dx = (x_pan[-1] - x_pan[0]) / (len(x_sub) - 1)
        shaded_area += np.sum(0.5 * (np.abs(y_sub[:-1]) + np.abs(y_sub[1:])) * dx)

        if plot:
            fill_label = 'Parabolic Interpolant' if i == 0 else None
            plt.fill_between(x_sub, 0, y_sub, color='skyblue', edgecolor='navy', alpha=0.4, linewidth=1.2)
            plt.plot(x_sub, y_sub, color='crimson', linewidth=1.5, label=fill_label)

    if plot:
        plt.plot(x_pts, fx, 'ro', markersize=5, label='Evaluation Nodes')
        plt.axhline(0, color='black', linewidth=0.8)
        plt.title(f"Simpson's 1/3 Rule \n (Estimated Area = {shaded_area:0.4f})")
        plt.xlabel('x')
        plt.ylabel('y')
        plt.legend(loc='upper right')
        plt.grid(True, ls=':', alpha=0.4)

    return {
        'integral': integral,
        'shaded_area': shaded_area,
        'x_pts': x_pts,
        'fx': fx
    }
```

Similarly to the Trapezoidal rule, we divide the interval into n equally spaced evaluate nodes. We estimate the integral by using a characteristic weighting array multiplied by a step factor of h/3.

```python
def simpson_cubic(func, a, b, n, plot=True):
    if (n - 1) % 3 != 0:
        n += 3 - ((n - 1) % 3)
    
    h = (b - a) / (n - 1)
    x_pts= a + np.arange(n) * h
    fx = func(x_pts)
    
    cs = 3 * np.ones(n); cs[0::3] = 2; cs[0] = 1; cs[-1] = 1
    contribs = cs*func(x_pts)
    integral = (3 * h / 8) * np.sum(contribs)

    num_panels = (n - 1) // 3
    shaded_area = 0.0

    if plot:
         x_dense = np.linspace(a, b, 500)
         plt.plot(x_dense, func(x_dense), 'k--', label='$f(x)$ True Curve', alpha=0.7)
        
    for i in range(num_panels):
        idx = [3 * i, 3 * i + 1, 3 * i + 2, 3 * i + 3]

        x_pan = x_pts[idx]
        y_pan = fx[idx]

        poly = np.polyfit(x_pan, y_pan, 3)

        x_sub = np.linspace(x_pan[0], x_pan[-1], 200)
        y_sub = np.polyval(poly, x_sub)

        dx = (x_pan[-1] - x_pan[0]) / (len(x_sub) - 1)
        shaded_area += np.sum(0.5 * (np.abs(y_sub[:-1]) + np.abs(y_sub[1:])) * dx)

        if plot:
            fill_label = 'Cubic Interpolant' if i == 0 else None
            plt.fill_between(x_sub, 0, y_sub, color='skyblue', edgecolor='navy', alpha=0.4, linewidth=1.2)
            plt.plot(x_sub, y_sub, color='crimson', linewidth=1.5, label=fill_label)

    if plot:
        plt.plot(x_pts, fx, 'ro', markersize=5, label='Evaluation Nodes')
        plt.axhline(0, color='black', linewidth=0.8)
        plt.title(f"Simpson's 3/8 Rule \n (Estimated Area = {shaded_area:0.4f})")
        plt.xlabel('x')
        plt.ylabel('y')
        plt.legend(loc='upper right')
        plt.grid(True, ls=':', alpha=0.4)

    return {
        'integral': integral,
        'shaded_area': shaded_area,
        'x_pts': x_pts,
        'fx': fx
    }
```

Here, the 3/8 rule works similarly to the 1/3 rule, only with different weighting and scaling factor. Additionally, the code automatically adjusts the number of partitions into a number divisible by 3.
