### Problem 2: Simple Harmonic Oscillator

The position $x(t)$ of a 1D simple harmonic oscillator is described by the equation of motion:

$$-kx = \frac{d^2x}{dt^2}$$

We know a solution can take the form of $x(t) = \cos(t)$. Calculate the velocity of this particle using any finite-difference method, and plot its absolute error with respect to the analytical value by varying your step-sizes. The absolute error should have the same trend as the $\mathcal{O}$ error of your chosen method.

```python
import numpy as np
import matplotlib.pyplot as plt

# Position
def x(t):
    return np.cos(t)

# Analytical solution
def v_exact(t):
    return -np.sin(t)

# Forward difference
def fordiff(t, h):
    return (x(t+h) - x(t))/h

# Central difference
def centraldiff(t, h):
    return (x(t+h/2) - x(t-h/2))/h

# Backward difference
def backdiff(t, h):
    return (x(t) - x(t-h))/h 


# Set up the calculation
t0 = 1.0
v_true = v_exact(t0)

# Different positive step sizes
h_values = np.linspace(0.00001, 2, 500)

# Calculate numerical derivatives
v_fd = fordiff(t0, h_values)
v_cd = centraldiff(t0, h_values)
v_bd = backdiff(t0, h_values)

# Calculate absolute errors
error_fd = np.abs(v_fd - v_true)
error_cd = np.abs(v_cd - v_true)
error_bd = np.abs(v_bd - v_true)


# Plot absolute error
plt.figure(figsize=(8, 5))

plt.plot(h_values, error_fd, '-', markersize=3,
         color='tab:orange', label="Forward Difference")
plt.plot(h_values, error_cd, '-', markersize=3,
         color='tab:cyan', label="Central Difference")
plt.plot(h_values, error_bd, '-', markersize=3,
         color='tab:pink', label="Backward Difference")

# Reference slopes
plt.plot(h_values, 0.4*h_values, 'k--',
         alpha=0.7, label='O(h) Reference Slope')
plt.plot(h_values, 0.03*(h_values)**2, 'k:',
         alpha=0.7, label='O(h²) Reference Slope')

plt.xlabel("Step size, h")
plt.ylabel("Absolute error")
plt.title("Finite-Difference Absolute Error")
plt.grid(True, which="both", linestyle="--", alpha=0.6)
plt.legend()

plt.show()
```

At a fixed time $t_0 = 1.0$, the code calculates the numerical velocities for 500 different step sizes $h$ ranging from $0.00001$ to $2$. The absolute errors between the numerical and analytical velocities are then calculated and plotted against the step size, together with $\mathcal{O}(h)$ and $\mathcal{O}(h^2)$ reference slopes to compare the observed error trends with the expected accuracy of each finite-difference method.

### Derivatives

The derivate of a function $f(x) $ can be approximated using the **forward difference** as

$$f'(x) = \frac{f(x+h) - f(x)}{h} + \mathcal{O}(h)$$

using the **backward difference**

$$f'(x) = \frac{f(x) - f(x-h)}{h} + \mathcal{O}(h)$$

and the **central difference**

$$f'(x) = \frac{f(x + \frac{h}{2}) - f(x - \frac{h}{2})}{h} + \mathcal{O}(h^2)$$

where $h$ is a small step-size. To minimzie error, you can use Richardson Extrapolation, which can be found on the book (Gezerlis).
