### Problem 1: Finite Difference

Let $$f(x) = 1+\frac{1}{2}\tanh(2x)$$

**(a)** Create functions that take as input an array or list of points corresponding to a mathematical function and output the derivative using **(a.1)** forward-difference, **(a.2)** backward-difference, and **(a.3)** central-difference methods.


```python
import numpy as np
import matplotlib.pyplot as plt

# a.1

def forward_diff(x, y):
    h = x[1] - x[0] # (delta x) access second element and subtract the first element of the list
    dy = (y[1:] - y[:-1]) / h # (delta y over delta x) slices the first element and all succeeding elements and subtracts the following y values finding their difference, after which h is divided to approximate the slope
    return x[:-1], dy # last element is dropped to match length

# a.2

def backward_diff(x, y):
    h = x[1] - x[0]
    dy = (y[1:] - y[:-1]) / h 
    return x[1:], dy # performs the same thing as a.1 but backward
    
# a.3

def central_diff(x, y):
    h = x[1] - x[0]
    dy = (y[2:] - y[:-2]) / (2 * h) # calculated forward and backward steps and find their difference
    return x[1:-1], dy # first and last elements are dropped
```

**(b)** Calculate an analytic formula for the derivative. Make three subplots for the derivative of $f'(x)$, one for each method and the analytic answer on the same plots. Plot the exact answer (use `numpy`) as lines and the numerical one as dots. Describe how the two results compare. Use `plt.subplots`

The analytic derivative of $tanh(x)$ is $sech^2$. Hence:
$$\frac{1}{2} * 2 * sech^2(2x)$$
Which can be simplified to:
$$sech^2(2x) = 1 - tanh^2(2x)$$

```python
def f(x):
    return 1 + (1/2) * np.tanh(2*x)

def analytic_f(x):
    return 1.0 / (np.cosh(2*x))**2 # wala pala sech sa numpy so we use cosh instead

# generating linspace to be used in our equations in part A, these are our sample data points
x_grid = np.linspace(-1, 1, 30)
y_grid = f(x_grid)

# perfomring each type
x_fwrd, dy_fwrd = forward_diff(x_grid, y_grid)
x_cntr, dy_cntr = central_diff(x_grid, y_grid)
x_bwrd, dy_bwrd = backward_diff(x_grid, y_grid)

fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True) #sharey aligns plots so they have the same scale
x_dense = np.linspace(-1, 1, 200) # grid size, higher = smoother

# Forward Difference
axes[0].plot(x_dense, analytic_f(x_dense), 'r-', label='Analytic') # plots the analytic function
axes[0].plot(x_fwrd, dy_fwrd, 'bo', label='Forward') # plots the 
axes[0].set_title("Forward Difference")
axes[0].legend()
axes[0].grid(True)

# Central Difference
axes[1].plot(x_dense, analytic_f(x_dense), 'g-', label='Analytic')
axes[1].plot(x_cntr, dy_cntr, 'bo', label='Center')
axes[1].set_title("Central Difference")
axes[1].legend()
axes[1].grid(True)

# Backward Difference
axes[2].plot(x_dense, analytic_f(x_dense), 'r-', label='Analytic')
axes[2].plot(x_bwrd, dy_bwrd, 'bo', label='Backward')
axes[2].set_title("Backward Difference")
axes[2].legend()
axes[2].grid(True)

plt.tight_layout()
plt.savefig('Difference_Comparison.png', dpi=150)
plt.show()
```

### Problem 2: Simple Harmonic Oscillator

The position $x(t)$ of a 1D simple harmonic oscillator is described by the equation of motion:

$$-kx = \frac{d^2x}{dt^2}$$

We know a solution can take the form of $x(t) = \cos(t)$. Calculate the velocity of this particle using any finite-difference method, and plot its absolute error with respect to the analytical value by varying your step-sizes. The absolute error should have the same trend as the $\mathcal{O}$ error of your chosen method.

#### Solution

Position $x(t)$ can be used to obtain Velocity $v(t)$ by taking it's first derivative. Similarly Acceleration $a(t)$ can also be obtained by taking the second derivative of $x$. We know that $x(t)$ is usually in the form of a sin or a cos wave.

```python
# Defining a grid of time with dt let's say 10 seconds
t = np.linspace(0, 10, 500)
dt = t[1] - t[0]

# defining x(t), this assumed a theoretical position for x(t) using cos
w = 2.0 # radians per second
x = np.cos(w * t) # this is basically the x(t) = cos(t)

# analytic derivative of x is -sin(t)
# dx = -1 * np.sin(t) wait we don't really need to do this because we can just take the central difference :O

# I will be using central difference

# taking the derivative of this, v = dx/dt
v = (x[2:] - x[:-2]) / (2 * dt)
t_v = t[1:-1] # aligning

# taking 2nd derivative of x using v, a = dv/dt
a = (v[2:] - v[:-2]) / (2 * dt)
t_a = t[2:-2] # aligning




################################################################ absolute error



# target time
t_target = 1
# we first take the analytic derivative of v
v_analytic = -w * np.sin(w*t_target)
# calculating absolute error
v_error = np.abs(v - v_analytic)

# Generate an array of step sizes (h)
h_values = np.logspace(-4, 0, 100)
errors = []

for h in h_values:
    # Central difference points around t_target
    x_minus = np.cos(w * (t_target - h))
    x_plus = np.cos(w * (t_target + h))
    
    # Numerical derivative of v
    v_numerical = (x_plus - x_minus) / (2 * h)
    
    # Calculate absolute error
    error = np.abs(v_numerical - v_analytic)
    errors.append(error)




############################################################################




# Figure Plot
plt.figure(figsize=(15, 6))
plt.plot(t, x, label='Position $x(t)$', color='blue', linewidth=2)
plt.plot(t_v, v, label='Velocity $v(t)$', color='red', linestyle='--', linewidth=2)
plt.plot(t_a, a, label='Acceleration $a(t)$', color='orange', linestyle=':', linewidth=1)

plt.title('Simple Harmonic Oscillator Using Central Difference')
plt.xlabel('Time ($s$)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig('SHO.png', dpi=150)
plt.show()

# Error Plot
plt.figure(figsize=(15, 5))
plt.plot(h_values, errors, label='Numerical Error', color='red', linewidth=2)

plt.title('Error Scaling vs. Step Size $h$ at $t=1$')
plt.xlabel('Step Size $h$')
plt.ylabel('Absolute Error')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig('ErrorScaling_vs_StepSize.png', dpi=150)
plt.show()
```

### Derivatives

The derivate of a function $f(x) $ can be approximated using the **forward difference** as

$$f'(x) = \frac{f(x+h) - f(x)}{h} + \mathcal{O}(h)$$

using the **backward difference**

$$f'(x) = \frac{f(x) - f(x-h)}{h} + \mathcal{O}(h)$$

and the **central difference**

$$f'(x) = \frac{f(x + \frac{h}{2}) - f(x - \frac{h}{2})}{h} + \mathcal{O}(h^2)$$

where $h$ is a small step-size. To minimzie error, you can use Richardson Extrapolation, which can be found on the book (Gezerlis).
