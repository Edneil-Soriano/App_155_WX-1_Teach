All three finite-difference methods closely approximate the analytic derivative $f'(x) = \operatorname{sech}^2(2x)$ across the tested range, with the numerical dots tracking the analytic curve almost exactly at $h = 0.1$. This confirms that forward, backward, and central differences all correctly capture the shape of the derivative, a sharp peak near $x=0$ where $\tanh(2x)$ transitions fastest, decaying smoothly to zero as $|x|$ grows and the function flattens out.

On closer inspection, the forward-difference curve appears slightly shifted to the left of the analytic peak, while the backward-difference curve appears slightly shifted to the right. This happens because forward difference estimates the slope using a point *ahead* of $x$ (at $x+h$), so it effectively reports the derivative "early," while backward difference uses a point *behind* $x$ (at $x-h$), reporting it "late", both are really approximating the slope at the midpoint of their sampling interval rather than exactly at $x$. Central difference avoids this shift entirely by sampling symmetrically around $x$ (at $x\pm h/2$), which is also why it lines up most accurately with the analytic curve.

The shapes trace back to each method's leading Taylor-series error term. Forward and backward difference are $O(h)$, with leading error $\frac{h}{2}x''(t_0) = -\frac{h}{2}\cos(t_0)$ — since this is *linear* in $h$, their error grows roughly as a straight line for small-to-moderate $h$, bending slightly at larger $h$ (near 2, a sizable fraction of the period $2\pi$) as higher-order terms start contributing.

Central difference is $O(h^2)$, with leading error $\frac{h^2}{24}x'''(t_0) = \frac{h^2}{24}\sin(t_0)$. Being *quadratic* in $h$, its error stays near zero for small $h$ and only curves upward noticeably at larger $h$, explaining the flatter blue curve compared to the steeper orange/green lines.

The differing magnitudes across Figures 2–4 come from these coefficients depending on $\cos(t_0)$ and $\sin(t_0)$. At $t_0=0.5$, $\cos(0.5)\approx 0.88$ is large, so forward/backward error dominates sharply (Figure 2). At $t_0=5$, $\sin(5)\approx -0.96$ dominates while $\cos(5)\approx 0.28$ is small, making central difference's error more visible and forward difference overtake backward difference (Figure 4).

The dips toward zero occur when large $h$ causes the approximation to momentarily land close to the true oscillating value by coincidence, not genuine convergence.

### Problem 1: Finite Difference

Let $$f(x) = 1+\frac{1}{2}\tanh(2x)$$

**(a)** Create functions that take as input an array or list of points corresponding to a mathematical function and output the derivative using **(a.1)** forward-difference, **(a.2)** backward-difference, and **(a.3)** central-difference methods.


```python
import numpy as np

def f(x):
    return 1 + (np.tanh(2*x))/2

#a.1 forward difference
def forward_dif(f, x, h):
    fd = (f(x+h)- f(x))/h
    return fd

#a.2 backward difference
def backward_dif(f, x, h):
    bd = (f(x)- f(x-h))/h
    return bd

#a.3 central difference
def central_dif(f, x, h):
    cd = (f(x+h/2) - f(x-h/2))/h
    return cd
```

**(b)** Calculate an analytic formula for the derivative. Make three subplots for the derivative of $f'(x)$, one for each method and the analytic answer on the same plots. Plot the exact answer (use `numpy`) as lines and the numerical one as dots. Describe how the two results compare. Use `plt.subplots`

```python
import numpy as np
import matplotlib.pyplot as plt

def fprime(x):
    return 1/((np.cosh(2*x))**2)

x = np.linspace(-3, 3, 200)
h = 0.1

plt.figure(figsize=(15,4))

# plotting forward difference
plt.subplot(1,3,1)
plt.plot(x, fprime(x), label="Analytic")
plt.plot(x, forward_dif(f,x,h), '.', label="Numerical")
plt.title("Forward Difference")
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.grid()
plt.legend()

# plotting backward difference
plt.subplot(1,3,2)
plt.plot(x, fprime(x), label="Analytic")
plt.plot(x, backward_dif(f,x,h), '.', label="Numerical")
plt.title("Backward Difference")
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.grid()
plt.legend()

# plotting central difference
plt.subplot(1,3,3)
plt.plot(x, fprime(x), label="Analytic")
plt.plot(x, central_dif(f,x,h), '.', label="Numerical")
plt.title("Central Difference")
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.grid()
plt.savefig('fig1.png', format='png', dpi=300)
plt.legend()
```

#### Probem 1 Code Summary

This code defines each finite-difference method as its own reusable function so all three (forward, backward, central) can compute derivatives from the same `f(x)` and `x` array, then compares each against the true analytic derivative `fprime(x)` on a shared set of `plt.subplot(1,3,i)` panels so the visual comparison is fair, same x-range, same `h`, same axes scale. Plotting the analytic result as a line and each numerical result as dots (`'.'`) makes it easy to see how closely each method tracks the true derivative, with the three side-by-side panels showing that all three converge well at `h=0.1`, though central difference should sit closest to the analytic line since it's the more accurate O(h²) method.


### Problem 2: Simple Harmonic Oscillator

The position $x(t)$ of a 1D simple harmonic oscillator is described by the equation of motion:

$$-kx = \frac{d^2x}{dt^2}$$

We know a solution can take the form of $x(t) = \cos(t)$. Calculate the velocity of this particle using any finite-difference method, and plot its absolute error with respect to the analytical value by varying your step-sizes. The absolute error should have the same trend as the $\mathcal{O}$ error of your chosen method.

```python
import math
import matplotlib.pyplot as plt

# position function
def x(t):
    return math.cos(t)

# analytical velocity: dx/dt= -sin(t)
def analytical_v(t):
    return -math.sin(t)

# using central difference for velocity
def central_dif(f,t,h):
    return (f(t+h/2) - f(t-h/2))/h

# using t0 as some fixed point to test
t0 = 0.5

hs = np.linspace(0.01, 2, 100)
errors_central = []
for h in hs:
    numerical_v = central_dif(x, t0, h)
    error = abs(numerical_v - analytical_v(t0))
    errors_central.append(error)

errors_backwards = []
for h in hs:
    numerical_v = backward_dif(x, t0, h)
    error = abs(numerical_v - analytical_v(t0))
    errors_backwards.append(error)

errors_forward = []
for h in hs:
    numerical_v = forward_dif(x, t0, h)
    error = abs(numerical_v - analytical_v(t0))
    errors_forward.append(error)

# plot absolute error vs step size

plt.plot(hs, errors_central, label="Central Difference")
plt.plot(hs, errors_backwards, label="Backward Difference")
plt.plot(hs, errors_forward, label="Forward Difference")
plt.xlabel("Step Size (h)")
plt.ylabel("Absolute Error")
plt.title("Central Difference Absolute Error with t0=0.5")
plt.legend()
plt.grid()

plt.savefig('fig2.png', format='png', dpi=300) 
plt.show()

```

#### Problem 2 code summary

This code tests all three finite-difference methods (central, backward, forward) against the same fixed point t0 and the same range of step sizes hs, computing each one's absolute error against the known analytical velocity so the three methods are compared under identical, controlled conditions. It then plots all three error curves on one graph and a legend, gridlines, and saves the figure. This visually confirms that $O(h^2)$ converges faster than the $O(h)$ forward and backward methods as h shrinks.

### Derivatives

The derivate of a function $f(x) $ can be approximated using the **forward difference** as

$$f'(x) = \frac{f(x+h) - f(x)}{h} + \mathcal{O}(h)$$

using the **backward difference**

$$f'(x) = \frac{f(x) - f(x-h)}{h} + \mathcal{O}(h)$$

and the **central difference**

$$f'(x) = \frac{f(x + \frac{h}{2}) - f(x - \frac{h}{2})}{h} + \mathcal{O}(h^2)$$

where $h$ is a small step-size. To minimzie error, you can use Richardson Extrapolation, which can be found on the book (Gezerlis).
