### Problem 1 Finite Difference

![Finite-Difference Approximations of the Derivative vs. Analytical Solution](4-problem_1.png)

Figure 1. Finite-Difference Approximations of the Derivative vs. Analytical Solution

Figure 1 compares three ways of estimating f'(x) against the exact analytical answer. The forward-difference points sit a bit above the curve on the left side (x < 0) and a bit below it on the right side (x > 0), while the backward-difference points do the opposite. This happens because forward difference uses the point just ahead of x, so it's slightly biased toward wherever the curve is heading next; backward difference uses the point just behind x, so it's biased toward where the curve just came from. Since the bump curves upward before the peak and downward after it, this bias flips direction right at x = 0. Central difference uses points on both sides equally, so these two biases cancel out. That's why its dots line up almost exactly with the analytical curve, with much less error than either one-sided method.

### Problem 1: Finite Difference

Let $$f(x) = 1+\frac{1}{2}\tanh(2x)$$

**(a)** Create functions that take as input an array or list of points corresponding to a mathematical function and output the derivative using **(a.1)** forward-difference, **(a.2)** backward-difference, and **(a.3)** central-difference methods.


```python
# code here
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return ( 1 + 0.5 * np.tanh(2*x))

# analytic derivative f'(x); sech^2(2x) = 1-tanh^2(2x)
def f_prime(x):
    return (1 - np.tanh(2*x)**2)

# forward difference
def forward_diff(f, x, h):
    return ((f(x+h) -f(x)) / h)
    
# backward difference
def backward_diff(f, x, h):
    return ((f(x)-f(x-h))/h)
    
# central difference
def central_diff(f, x, h):
    return ((f(x + h/2) - f(x - h/2)) / (h))

#setup
x = np.linspace(-3,3, 100) # domain to evaluate over, from -3 to 3 with 100 points
h = x[1] - x[0] # uniform grid spacing, used as the step size
y = f(x)

y_exact = f_prime(x)
y_forward = forward_diff(f, x, h)
y_backward = backward_diff(f, x,h)
y_central = central_diff(f, x,h)
```

**(b)** Calculate an analytic formula for the derivative. Make three subplots for the derivative of $f'(x)$, one for each method and the analytic answer on the same plots. Plot the exact answer (use `numpy`) as lines and the numerical one as dots. Describe how the two results compare. Use `plt.subplots`

```python
#plotting
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15,4))

ax1.plot(x, y_exact, 'k-', label="Analytic", linewidth = 2)
ax1.plot(x, y_forward, 'o', color='tab:blue', label='Numerical', markersize=4)
ax1.set_title('Forward Difference')
ax1.set_xlabel('x')
ax1.set_ylabel("f'(x)")
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend()

ax2.plot(x, y_exact, 'k-', label="Analytic", linewidth=2)
ax2.plot(x, y_backward, 'o', color='tab:green', label = "Numerical", markersize=4)
ax2.set_title("Backward Difference")
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.grid(True, linestyle='--', alpha=0.6)
ax2.legend()

ax3.plot(x, y_exact, 'k-', label="Analytic", linewidth=2)
ax3.plot(x, y_central, 'o', color='tab:pink', label="Numerical", markersize=4)
ax3.set_title("Central Difference")
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.grid(True, linestyle='--', alpha=0.6)
ax3.legend()

fig.suptitle("Finite-Difference Approximations of the Derivative vs. Analytical Solution", fontsize=18)
plt.tight_layout()
plt.savefig('4-problem_1.png')
plt.show()
```

* This code evaluates three different numerical differentiation methods, using three separate functions, then compares their outputs against the exact analytical derivative across a grid of points. The step size h is taken directly from the grid spacing (x[1] - x[0]) so the approximation's resolution matches the resolution of the plotted grid, rather than picking an arbitrary separate value. It was plotted in three side-by-side subplots for easy visual comparison of their differences.

### Problem 2: Simple Harmonic Oscillator

The position $x(t)$ of a 1D simple harmonic oscillator is described by the equation of motion:

$$-kx = \frac{d^2x}{dt^2}$$

We know a solution can take the form of $x(t) = \cos(t)$. Calculate the velocity of this particle using any finite-difference method, and plot its absolute error with respect to the analytical value by varying your step-sizes. The absolute error should have the same trend as the $\mathcal{O}$ error of your chosen method.

```python
import matplotlib.pyplot as plt
import numpy as np

# position
def x(t):        
    return np.cos(t)
    
# velocity analytical solution
def v_exact(t):
    return -np.sin(t)

# range of step sizes to test
h_values = np.linspace(0.0001, 2, 100)
t0_values = [0, 1, 2] # evaluation points, one per subplot

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

for ax, t0 in zip(axes, t0_values):
    v_true = v_exact(t0)
    
    # calculate absolute errors
    error_backward = [np.abs(backward_diff(x, t0, h) - v_true) for h in h_values]
    error_central  = [np.abs(central_diff(x, t0, h) - v_true) for h in h_values]

    # plotting
    ax.plot(h_values, error_backward, '-', label="Backward Difference", color='tab:cyan')
    ax.plot(h_values, error_central, '-', label="Central Difference", color='tab:pink')
    ax.plot(h_values, 0.4*h_values, 'k--', alpha=0.7, label='O(h) Reference Slope')
    ax.plot(h_values, 0.03*h_values**2, 'k:', alpha=0.7, label='O(h^2) Reference Slope')

    ax.set_title(f"t0 = {t0}")
    ax.set_xlabel("Step size (h)")
    ax.set_ylabel("Absolute Error")
    ax.grid(True)
    ax.legend(fontsize=8)

plt.suptitle("Error of Numerical Differentiation vs. Analytical Solution at t=t0", fontsize=18)
plt.tight_layout()
plt.savefig('4-problem_2.png')
plt.show()
```

### Derivatives

The derivate of a function $f(x) $ can be approximated using the **forward difference** as

$$f'(x) = \frac{f(x+h) - f(x)}{h} + \mathcal{O}(h)$$

using the **backward difference**

$$f'(x) = \frac{f(x) - f(x-h)}{h} + \mathcal{O}(h)$$

and the **central difference**

$$f'(x) = \frac{f(x + \frac{h}{2}) - f(x - \frac{h}{2})}{h} + \mathcal{O}(h^2)$$

where $h$ is a small step-size. To minimzie error, you can use Richardson Extrapolation, which can be found on the book (Gezerlis).
