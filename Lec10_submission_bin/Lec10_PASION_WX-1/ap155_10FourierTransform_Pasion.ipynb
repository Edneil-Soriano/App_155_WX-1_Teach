### Problem 1: Exploring FT of varying waveforms

The shape of a waveform determines how its energy is distributed across the frequency domain. Sharp transitions, such as those in a square wave, produce multiple higher-frequency harmonics, while smoother waveforms, such as triangle waves, exhibit a faster decrease in harmonic amplitudes. For a Gaussian waveform, increasing its standard deviation ($\sigma$) broadens the signal in the time domain, resulting in a narrower Fourier transform, whereas decreasing $\sigma$ produces a narrower pulse and a broader frequency spectrum. This demonstrates the inverse relationship between the spread of a signal in the time domain and its frequency domain representation.

### Problem 2: Signal Processing in 1D

Reducing $N$ shortens the duration of the sampled signal while keeping the sampling frequency fixed at $5000\text{ Hz}$. This means fewer cycles of the original sinusoidal components are captured, making their frequency peaks less distinguishable from the noise in the Fourier spectrum. As $N$ decreases (particularly around $N = 5000$), signal recovery becomes less accurate because noise peaks can exceed the filtering threshold and reappear in the reconstructed signal, increasing distortion and RMS error. There is therefore a practical lower limit for reliable reconstruction (approximately $N = 10000$ in this case), since the amplitudes of coherent signal components scale with $N$, whereas the typical magnitude of uncorrelated Gaussian noise scales with $\sqrt{N}$.

### Problem 3: Aliasing

The $9\text{ Hz}$ sinusoidal signal was sampled at $100\text{ Hz}$ and $12\text{ Hz}$ to demonstrate the effects of sampling above and below the Nyquist frequency, respectively. At $f_s = 100\text{ Hz}$, the sampling rate satisfies the Nyquist criterion, preserving the original waveform and producing a Fourier transform peak at $9\text{ Hz}$. In contrast, sampling at $f_s = 12\text{ Hz}$ results in aliasing, causing the reconstructed waveform to appear as a $3\text{ Hz}$ sinusoid and producing a corresponding peak at $3\text{ Hz}$ in the Fourier spectrum. This demonstrates how undersampling leads to an incorrect representation of the original signal's frequency.

Given the waveforms from the pool below, produce a $1\times3$ plot containing the following:

```python
# Defining the parameters
N = 1000
frequency = 2
amplitude = 1
mu = 5.0 # Mean (center) of the Gaussian curve in time
sigma = 1.0 # Standard deviation (controls the width/spread of the Gaussian curve)

# Define time axis
time_axis = np.linspace(0, 10, N)
deltaT = time_axis[1] - time_axis[0]

# Defining different Waveforms
square_wave = amplitude * signal.square(2 * np.pi * frequency * time_axis)
triangle_wave = amplitude * signal.sawtooth(2 * np.pi * frequency * time_axis, width = 0.5)
gaussian_wave = 1 / (sigma * np.sqrt(2*np.pi)) * np.exp(-((time_axis - mu) ** 2) / (2 * (sigma**2)))
```

```python
# Square Wave

sw_ft = np.fft.fft(square_wave)
freqs = np.fft.fftfreq(N, deltaT)

"""
# Q: Why do fftshift?
fftshift rearranges the FFT output so that the zero-frequency
component is at the center, with negative frequencies on the left
and positive frequencies on the right.
"""

freqs = np.fft.fftshift(freqs) 
sw_ft = np.fft.fftshift(sw_ft) 

"""
# Q: Why unwrap?
Unwrapping removes sudden 2pi jumps in the phase by adding or
subtracting multiples of 2pi, making the phase plot continuous.
"""

sw_mag = np.abs(sw_ft)
sw_phase = np.angle(sw_ft)
sw_phase = np.unwrap(sw_phase) 

fig, ax = plt.subplots(1,3, figsize=(12,3))

# Plotting the signal
ax[0].set_title("Square Wave, f = 2 [Hz]")
ax[0].set_ylabel("Amplitude") 
ax[0].set_xlabel("Time (t)")
ax[0].plot(time_axis, square_wave, color="blue")
ax[0].grid(True)

# Plotting the magnitude spectrum
ax[1].set_title("FFT (Magnitude)")
ax[1].set_xlabel("Frequency (f)")
ax[1].set_ylabel("Magnitude") 
ax[1].plot(freqs, sw_mag, color="green")
ax[1].set_xlim(-4, 4)
ax[1].grid(True)

# Plotting the phase spectrum
ax[2].set_title("FFT (Phase)")
ax[2].set_xlabel("Frequency (f)")
ax[2].set_ylabel("Phase (radians)") 
ax[2].plot(freqs, sw_phase, color="red")
ax[2].set_xlim(-4, 4)
ax[2].grid(True)

fig.tight_layout()
plt.savefig("Figures/e10_figure0", dpi=300)
plt.show()
```

```python
# Triangle Wave

freqs = np.fft.fftfreq(N, deltaT)
tw_ft = np.fft.fft(triangle_wave)

freqs = np.fft.fftshift(freqs)
tw_ft = np.fft.fftshift(tw_ft)

tw_mag = np.abs(tw_ft)
tw_phase = np.unwrap(np.angle(tw_ft))

fig, ax = plt.subplots(1, 3, figsize=(12, 3))

# Plotting signal
ax[0].set_title("Triangle Wave, f = 2 [Hz]")
ax[0].set_xlabel("Time (t)")
ax[0].set_ylabel("Amplitude")
ax[0].plot(time_axis, triangle_wave, color="blue")
ax[0].grid(True)

# Plotting magnitude spectrum
ax[1].set_title("FFT (Magnitude)")
ax[1].set_xlabel("Frequency (f)")
ax[1].set_ylabel("Magnitude")
ax[1].plot(freqs, tw_mag, color="green")
ax[1].set_xlim(-4, 4)
ax[1].grid(True)

# Plotting phase spectrum
ax[2].set_title("FFT (Phase)")
ax[2].set_xlabel("Frequency (f)")
ax[2].set_ylabel("Phase (radians)")
ax[2].plot(freqs, tw_phase, color="red")
ax[2].set_xlim(-4, 4)
ax[2].grid(True)

fig.tight_layout()
plt.savefig("Figures/e10_figure1.png", dpi=300)
plt.show()
```

```python
# Gaussian function

freqs = np.fft.fftfreq(N, deltaT)
gw_ft = np.fft.fft(gaussian_wave)

freqs = np.fft.fftshift(freqs) 
gw_ft = np.fft.fftshift(gw_ft)

gw_mag = np.abs(gw_ft)
gw_phase = np.angle(gw_mag)
gw_phase = np.unwrap(gw_phase)

pos = freqs >= 0 # Select nonnegative frequencies

fig, ax = plt.subplots(1,3, figsize=(12,3))

ax[0].set_title("Gaussian Wave, Center = 5")
ax[0].set_ylabel("Amplitude")
ax[0].set_xlabel("Time (t)")
ax[0].plot(time_axis, gaussian_wave, color="blue")
ax[0].grid(True)

ax[1].set_title("FFT (Magnitude)")
ax[1].set_ylabel("Magnitude")
ax[1].set_xlabel("Frequency (f)")
ax[1].plot(freqs[pos], gw_mag[pos], color="green")
ax[1].set_xlim(0, 4)
ax[1].grid(True)

ax[2].set_title("FFT (Phase)")
ax[2].set_ylabel("Phase (radians)")
ax[2].set_xlabel("Frequency (f)")
ax[2].plot(freqs[pos], gw_phase[pos], color="red")
ax[2].set_xlim(0, 4)
ax[2].grid(True)

fig.tight_layout()
plt.savefig("Figures/e10_figure2", dpi=300)
plt.show()
```

### Problem 3: Aliasing
1) Using different sampling frequencies (uniform sampling), generate a sinusoidal signal. Note the sampling frequencies, and the signal frequency. Use a sampling frequency above the Nyquist frequency, and another one below it.
3) Get their fourier transforms. Are they the same?
4) Plot: Show the fourier transform of an aliased signal, how it looks like in the time-domain, and how the actual signal looks like in the time domain if it's not aliased.
   
Discussion: Explain the report figure. Note the frequency of the signal you wanted to sample, the sampling frequencies used for the aliased and the non-aliased wave

```python
f = 9
T = 10
fs_list = [100, 12]  # Sampling rates: above and below the Nyquist rate

t_fine = np.linspace(0, T, 10000)
x_real = np.sin(2 * np.pi * f * t_fine)

for fs in fs_list:
    f_nyquist = 2 * f
    if fs > f_nyquist:
        nyq_label = "Above Nyquist frequency (No Aliasing)"
    else:
        nyq_label = "Below Nyquist frequency (Aliased)"

    N = int(T * fs)
    t = np.linspace(0, T, N, endpoint=False)
    deltaT = t[1] - t[0]

    x = np.sin(2 * np.pi * f * t)

    freqs = np.fft.fftfreq(N, deltaT)
    x_ft = np.fft.fft(x)

    freqs = np.fft.fftshift(freqs)
    x_ft = np.fft.fftshift(x_ft)

    x_mag = np.abs(x_ft)
    x_phase = np.angle(x_mag)
    x_phase = np.unwrap(x_phase)

    pos = freqs >= 0

    print(
        f"Sampling Rate: {fs} Hz ({nyq_label}): "
        f"dominant frequency = "
        f"{freqs[pos][np.argmax(x_mag[pos])]:.2f} Hz")

    fig, ax = plt.subplots(1, 3, figsize=(14, 3.5))

    fig.suptitle(
        f"fs = {fs} Hz | {nyq_label}",
        fontsize=13)

    ax[0].set_title("Original Sine Wave and Sampled Data")
    ax[0].set_xlabel("Time (s)")
    ax[0].set_ylabel("Amplitude")
    ax[0].plot(t_fine, x_real, color="lightgray",
               label="Original signal")
    ax[0].plot(t, x, ".-", label="Samples")
    ax[0].set_xlim(0, 1)
    ax[0].legend(fontsize=8)

    ax[1].set_title("FT Magnitude")
    ax[1].set_xlabel("Frequency (Hz)")
    ax[1].set_ylabel("FFT Magnitude")
    ax[1].plot(freqs[pos], x_mag[pos])
    ax[1].set_xlim(0, 15)

    ax[2].set_title("FT Phase")
    ax[2].set_xlabel("Frequency (Hz)")
    ax[2].set_ylabel("Phase (radians)")
    ax[2].plot(freqs[pos], x_phase[pos])

    fig.tight_layout()
    plt.savefig(f"Figures/e10_figure5_fs{fs}.png", dpi=300, bbox_inches="tight")
    plt.show()
```

### Fourier transform (from $t$-space to $\omega$-space)
$$ Y(\omega) = \int_{-\infty}^{+\infty} dt \frac{e^{-i\omega t}}{\sqrt{2\pi}}y(t)$$

### Inverse Fourier transform (from $\omega$-space to $t$-space)
$$ y(t) = \int_{-\infty}^{+\infty} d\omega Y(\omega) \frac{e^{i\omega t}}{\sqrt{2\pi}}$$

Something to think about: while we essentially do numerical integration, we do not impose h. Instead, we utilize the spacing of the dataset.

### Discrete FT/IFT

$$ Y_n = \frac{1}{\sqrt{2\pi}} \sum_{k=1}^N \left[(\cos(nk\theta)\Re\{y_k\}  +\sin(nk\theta)\Im\{y_k\} +i(\cos(nk\theta)\Im\{y_k\} −\sin(nk\theta)\Re\{y_k\})]\right]$$

$$ y_k = \frac{\sqrt{2\pi}}{N} \sum_{k=1}^N \left[(\cos(nk\theta)\Re\{Y_n\}  -\sin(nk\theta)\Im\{Y_n\} +i(\cos(nk\theta)\Im\{Y_n\} +\sin(nk\theta)\Re\{Y_n\})]\right]$$

Why bother it doing this way? 

It cleanly splits the real and imaginary part for both inputs and outputs, so that you can use a pair of floats rather than an aggregate complex data type (complex128 is just a pair of float64 anyway).

Using $Z=e^{−i\theta}, \rightarrow Z^{\pm nk} = e^{\mp ink\theta} = \cos(nk\theta)\mp i\sin(nk\theta)$.
