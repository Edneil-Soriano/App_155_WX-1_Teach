```python
# root finding
# fixed-point method

def gfunc(xs):
    return -xs**5 + 4*xs**4 -4*xs**3 + (xs**2)*(np.exp(xs)) -2*xs**2-4*xs*np.exp(xs) +9*xs +4*np.exp(xs) - 8
    #return (8-4*np.exp(xs))/(-xs**4 + 4*xs**3 - 4*xs**2 + xs*np.exp(xs)-2*xs-4*np.exp(xs)+8)




x_0 = -0.9999
n = x_0

for i in range(1):
    n = gfunc(n)

print(n)
```

```python
# bisection method

bs1 = 1.999999
bs2 = 2.000002

def ave(a,b):
    return (a+b)/2

bound1 = bs1
bound2 = bs2

for i in range(1000):
    if func(bound1)*func(ave(bound1,bound2)) < 0:
        bound1 = bound1
        bound2 = ave(bound1,bound2)
    
    else:
        if func(ave(bound1,bound2))*func(bound2) < 0:
            bound1 = ave(bound1, bound2)
            bound2 = bound2

        else:
            print("invalid bounds")
            break
        
print(bound1, bound2)
    
```

```python
# newton's method

x_0 = 0
dx = 0.1


def newton(dx, xes):
    return -func(xes)/((func(xes)+func(xes+dx))/dx)+xes

x_es = x_0
for i in range(1000):
    x_es = newton(dx, x_es)
    
print(x_es)
```

```python
# secant method

x_0 = 6
x_1 = 5


def secant(x_0, x_1):
    return x_0-func(x_0)*(x_0-x_1)/(func(x_0)-func(x_1))

x_es = x_0
x_ev = x_1
for i in range(1000):
    n = secant(dx, x_es)
    x_es = x_ev
    x_ev = n
    
print(n)
```
