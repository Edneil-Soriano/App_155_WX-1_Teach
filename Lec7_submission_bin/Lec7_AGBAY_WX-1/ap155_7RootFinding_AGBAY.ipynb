The convergence figure tracks the iterate sequence $x_i$ against step index $i$ across the function's three root regions, with horizontal dotted lines representing the theoretical roots of $f(x) = (x-2)^2(e^x - x^3 - 2)$. 

- Region 1 ($x \approx -1.1927$): For this simple root, Newton's method exhibits rapid quadratic convergence within 6 iterations. Secant overshoots initially due to slope approximation before settling superlinearly. Bisection reduces the interval at a steady, linear rate.
- Region 2 ($x = 2.0000$): This is a double root with multiplicity 2 ($f'(2) = 0$). Bisection fails completely because $(x-2)^2 \ge 0$ prevents any sign change ($f(a)f(b) < 0$). Newton's method degrades from quadratic to slow linear convergence due to the vanishing derivative, Secant oscillates erratically, and Fixed-Point iteration progresses well but slowly.
- Region 3 ($x \approx 4.5959$): Open methods (Newton and Secant) converge rapidly provided the initial guesses avoid overshooting back into Region 2. Fixed-point iteration using $g(x) = \ln(x^3 + 2)$ converges steadily, while bisection yields guaranteed linear reduction.

We now compare the four root-finding methods to analyze even further.

- Fixed-point iteration is easy to set up since it only needs a single guess and no derivative. However, it easily diverges or strays to the wrong root if your $x = g(x)$ setup fails the $\vert{}g'(x)\vert{} < 1$ condition near the root.
- Bisection is guaranteed to work whenever you have a continuous function and a valid bracket with a sign change. Its main drawbacks are that it converges slowly and completely fails on roots that touch the axis without crossing, like the double root at $x = 2$.
- Newton's method is usually the fastest option for simple roots, reaching high accuracy in just a few steps. Its main downsides are requiring an exact formula for the derivative and losing its speed advantage when dealing with multiple roots where the slope flattens to zero.
- The Secant method gives near-Newton speeds without needing an analytical derivative. The trade-off is that it requires two starting guesses and can jump erratically if consecutive function values are too close together.

To set up fixed-point iteration, we just rearrange $f(x) = 0$ into $x = g(x)$ so that any number solving this equation is a root. We turn this into a feedback loop by plugging our initial guess into $g(x)$ to get a new estimate, repeating the process to get closer to the answer.


By the fundamental theorem of algebra, since the highest degree of the function is 5, it would lead us to expect that the function has up to 5 roots. 

Note that the function can be rearranged and factored out to obtain the form:

$$
f(x) = (x - 2)^2 (e^x - x^3 - 2)
$$

To find the roots of $f(x) = (x-2)^2 (e^x - x^3 - 2)$ away from the double root at $x = 2$, set the second factor to zero:   

$$
e^x - x^3 - 2 = 0
$$

Isolate the exponential term $e^x$:

$$
e^x = x^3 + 2
$$

Take the natural logarithm ($\ln$) of both sides to isolate $x$:

$$
\ln(e^x) = \ln(x^3 + 2)
$$


$$
x = \ln(x^3 + 2)
$$

Whether this loop actually works depends on the derivative of $g(x)$ staying between negative one and positive one near the root so the errors get smaller each step. For our root, using $g(x) = \ln(x^3 + 2)$ satisfies this condition and cleanly pulls our guesses right to the solution.   

```python
import math
import matplotlib.pyplot as plt
import numpy as np

# Original function f(x)
def f(x):
    return -x**5 + 4*x**4 - 4*x**3 + (x**2)*math.exp(x) - 2*x**2 - 4*x*math.exp(x) + 8*x + 4*math.exp(x) - 8

# Analytical derivative f'(x)
def df(x):
    return -5*x**4 + 16*x**3 - 12*x**2 - 4*x + 8 + (x**2 - 2*x)*math.exp(x)

# Fixed-Point Iteration Functions g(x)
def g_A(x):
    # Isolating 8x -> g_A(x) (Used for Region 2 near x ≈ 2)
    return (x**5 - 4*x**4 + 4*x**3 - (x**2)*math.exp(x) + 2*x**2 + 4*x*math.exp(x) - 4*math.exp(x) + 8) / 8.0

def g_B(x):
    # Isolating e^x (x-2)^2 -> g_B(x) = ln(...) (Used for Region 3 near x ≈ 4.6)
    num = x**5 - 4*x**4 + 4*x**3 + 2*x**2 - 8*x + 8
    denom = (x - 2)**2
    val = num / denom if denom != 0 else 1e-12
    return math.log(val) if val > 0 else 0


# Generate x grid from -2 to 5 to inspect function behavior
x_vals = np.linspace(-2, 5, 1000)
y_vals = np.array([f(x) for x in x_vals])

plt.figure(figsize=(10, 5))
plt.plot(x_vals, y_vals, label=r"$f(x)$", color="blue", linewidth=2)
plt.axhline(0, color="black", linestyle="--", linewidth=1)

# Crop y-axis to observe axis crossings clearly
plt.ylim(-20, 20)
plt.xlim(-2, 5)

plt.title("Plot of $f(x)$ from $x = -2$ to $x = 5$ (Cropped Y-Axis)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.savefig("7plot.png")
plt.show()
```

```python
def fixed_point(g, x0, tol=1e-6, max_iter=100):
    history = [x0]
    x = x0
    for _ in range(max_iter):
        x_next = g(x)
        history.append(x_next)
        if abs(x_next - x) < tol:
            break
        x = x_next
    return history[-1], history
```

```python
def bisection(f, a, b, tol=1e-6, max_iter=100):
    history = []
    fa, fb = f(a), f(b)
    if fa * fb >= 0:
        print(f"Bisection failed on interval [{a}, {b}]: f(a) * f(b) >= 0 (no sign change).")
        return None, history
        
    for _ in range(max_iter):
        c = (a + b) / 2.0
        history.append(c)
        fc = f(c)
        if abs(fc) < tol or (b - a) / 2.0 < tol:
            break
        if fc * fa < 0:
            b, fb = c, fc
        else:
            a, fa = c, fc
    return history[-1], history
```

```python
def newton_raphson(f, df, x0, tol=1e-6, max_iter=100):
    history = [x0]
    x = x0
    for _ in range(max_iter):
        fx = f(x)
        dfx = df(x)
        if abs(fx) < tol:
            break
        if dfx == 0:
            print("Newton-Raphson failed: Derivative is zero.")
            break
        x = x - fx / dfx
        history.append(x)
    return history[-1], history
```

```python
def secant(f, x0, x1, tol=1e-6, max_iter=100):
    history = [x0, x1]
    for _ in range(max_iter):
        f0 = f(history[-2])
        f1 = f(history[-1])
        if abs(f1) < tol or abs(f1 - f0) < 1e-12:
            break
        x2 = history[-1] - f1 * (history[-1] - history[-2]) / (f1 - f0)
        history.append(x2)
    return history[-1], history
```

```python
# searched 3 different regions based on visual plot of f(x) to try to pin down the roots, I did this because I deemed it to be more interesting to see three separate plots
# given the interval of [-2,5]


# region 1: Interval [-2, 0], initial guesses around x = -1.0 
print("--- Region 1 ---")
r1_bis,  h1_bis  = bisection(f, a=-2.0, b=0.0)
r1_newt, h1_newt = newton_raphson(f, df, x0=-1.0)
r1_sec,  h1_sec  = secant(f, x0=-1.5, x1=-0.5)

# region 2: Interval [1, 3], initial guesses around x = 1.5 
print("\n--- Region 2 ---")
r2_bis,  h2_bis  = bisection(f, a=1.0, b=3.0)  # Fails due to no sign change
r2_fp,   h2_fp   = fixed_point(g_A, x0=1.9)
r2_newt, h2_newt = newton_raphson(f, df, x0=1.5)
r2_sec,  h2_sec  = secant(f, x0=1.5, x1=2.5)

# region 3: Interval [4, 5], initial guesses around x = 4.0 
print("\n--- Region 3 ---")
r3_bis,  h3_bis  = bisection(f, a=4.0, b=5.0)
r3_fp,   h3_fp   = fixed_point(g_B, x0=4.0)
r3_newt, h3_newt = newton_raphson(f, df, x0=4.8) 
r3_sec,  h3_sec  = secant(f, x0=4.0, x1=4.8)

print("\n--- Summary of Discovered Roots ---")
print(f"Region 1 Root: {r1_newt:.6f}")
print(f"Region 2 Root: {r2_newt:.6f}")
print(f"Region 3 Root: {r3_newt:.6f}")
```

```python
# plot plot plot
fig, axes = plt.subplots(3, 1, figsize=(10, 14))

# Region 1 Convergence
axes[0].plot(range(len(h1_bis)), h1_bis, 'o-', label="Bisection (a=-2.0, b=0.0)", color="blue", markersize=4)
axes[0].plot(range(len(h1_newt)), h1_newt, '^--', label="Newton-Raphson (x=-1.0)", color="red", markersize=5)
axes[0].plot(range(len(h1_sec)), h1_sec, 'd--', label="Secant (x=-1.5, x=-0.5)", color="purple", markersize=5)

# Expected theoretical root line
axes[0].axhline(y=-1.192686, color="black", linestyle=":", linewidth=1.5, label="Expected Root ($x \\approx -1.1927$)")

axes[0].set_title("Region 1 Convergence (Simple Root near $x \\approx -1.1927$)", fontsize=11, fontweight='bold')
axes[0].set_xlabel("Iteration Step ($i$)")
axes[0].set_ylabel("Approximated Value ($x_i$)")
axes[0].grid(True, linestyle=":", alpha=0.6)
axes[0].legend(loc="upper right")

# Region 2 Convergence (root with multiplicity 2 region where bisection failed)
if len(h2_fp) > 0:
    axes[1].plot(range(len(h2_fp)), h2_fp, 's-', label="Fixed-Point ($x$=1.9)", color="green", markersize=4)
axes[1].plot(range(len(h2_newt)), h2_newt, '^--', label="Newton-Raphson ($x$=1.5)", color="red", markersize=5)
axes[1].plot(range(len(h2_sec)), h2_sec, 'd--', label="Secant ($x$=1.5, $x$=2.5)", color="purple", markersize=5)

# Expected theoretical root line
axes[1].axhline(y=2.000000, color="black", linestyle=":", linewidth=1.5, label="Expected Root ($x = 2$, Multiplicity 2)")

axes[1].set_title("Region 2 Convergence (Double Root at $x = 2$)", fontsize=11, fontweight='bold')
axes[1].set_xlabel("Iteration Step ($i$)")
axes[1].set_ylabel("Approximated Value ($x_i$)")
axes[1].grid(True, linestyle=":", alpha=0.6)
axes[1].legend(loc="upper right")

# Region 3 Convergence
axes[2].plot(range(len(h3_bis)), h3_bis, 'o-', label="Bisection ($a$=4.0, $b$=5.0)", color="blue", markersize=4)
axes[2].plot(range(len(h3_fp)), h3_fp, 's-', label="Fixed-Point ($x$=4.0)", color="green", markersize=4)
axes[2].plot(range(len(h3_newt)), h3_newt, '^--', label="Newton-Raphson ($x$=4.8)", color="red", markersize=5)
axes[2].plot(range(len(h3_sec)), h3_sec, 'd--', label="Secant ($x$=4.0, $x$=4.8)", color="purple", markersize=5)

# Expected theoretical root line
axes[2].axhline(y=4.595864, color="black", linestyle=":", linewidth=1.5, label="Expected Root ($x \\approx 4.5959$)")

axes[2].set_title("Region 3 Convergence (Simple Root near $x \\approx 4.5959$)", fontsize=11, fontweight='bold')
axes[2].set_xlabel("Iteration Step ($i$)")
axes[2].set_ylabel("Approximated Value ($x_i$)")
axes[2].grid(True, linestyle=":", alpha=0.6)
axes[2].legend(loc="lower right")

plt.tight_layout()
plt.savefig("convergence_final.png")
plt.show()
```
